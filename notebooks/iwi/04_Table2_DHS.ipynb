# IWI experiment · Table 2 of the paper: wealth of the DHS clusters in PyTorch

This notebook re-runs the DHS experiment of *Analyzing Poverty through Intra-Annual Time-Series: A Wavelet Transform Approach* (Table 2) on the WaveletR1 data, with PyTorch in place of TensorFlow. The target is the International Wealth Index (IWI, 0–100) of each DHS cluster. The base model sees the 7 Landsat bands and the nighttime light (NL), and the other rows add NDVI or wavelet bands to it:

| key | Table 2 row | input bands | R² in the paper | Keras `val5` model on fold 5 |
|---|---|---|---|---|
| `base` | Base model using 7 raw Landsat images and NL data | 7 Landsat + NL | 0.81 | 0.78 |
| `base_ndvi` | Base model with temporal NDVI | base + 12 monthly NDVI | 0.83 | 0.75 |
| `base_wavelet6` | Base model and 6 statistical wavelet-based features | base + LFC and HFC minimum, maximum and mean | 0.83 | 0.78 |
| `base_hfc_max` | Base model and max values of HFC | base + HFC max | 0.81 | 0.76 |
| `base_hfc_mean` | Base model and mean values of HFC | base + HFC mean | 0.82 | 0.79 |
| `base_hfc_min` | Base model and min values of HFC | base + HFC min | 0.82 | 0.79 |
| `base_lfc_max` | Base model and max values of LFC | base + LFC max | 0.84 | 0.78 |
| `base_lfc_mean` | Base model and mean values of LFC | base + LFC mean | 0.82 | 0.80 |
| `base_lfc_min` | Base model and min values of LFC | base + LFC min | 0.81 | 0.79 |

**As in the paper and the `VGG16_temporal_simulation_*_DHS` notebooks**

* **Model and training.** As in 04: a 1 × 1 convolution with ReLU maps the bands to 3 channels, followed by the frozen ImageNet VGG-16 convolution layers, a flatten and one linear output with L2 0.01 on its weights. Training uses Adam with learning rate 1e-3, MSE loss, batch 64 and 100 epochs. The weights with the lowest validation loss are tested.
* **NL as a band.** The NL image of `NL.tif` (annual median radiance) is the eighth input band of every row.
* **Complete data only.** Clusters with a no-data pixel in any of their three files are left out, as in the paper (8,598 clusters there).
* **Folds.** Fold f is the test fold, the fold before it is the validation fold, and the other three are for training.

**Different from the paper**

* **Composites.** The paper used 3-year medians of Landsat 8 and VIIRS, and a 3-year median for every month of Landsat NDVI. The 01 files cover the survey year: the annual Landsat median (holes filled from the 3 years around it), the annual VIIRS median, and the MODIS NDVI of that year averaged to 12 months.
* **Clusters and folds.** All DHS clusters of `train_clusters_folds.csv` in VIIRS years (2014 and later, `NL_SENSOR`), with the spatial folds of 03.
* **Evaluation.** Each fold's model is tested on its own test fold only. The `*_DHS` notebooks test one model (`val5`: trained on folds 1–3, validated on fold 4) on all five folds, and the mean of their five values does not match Table 2 (LFC max: 0.79 against 0.84), so Table 2 probably comes from other runs. The last column above is that model on fold 5, the only fold it had not seen (for NDVI the notebook printed 25 of 26 batches).
* **Normalisation.** As in 04, the Landsat bands get the Collection 2 Level 2 scale factors of the GEE catalog: surface reflectance = DN × 0.0000275 − 0.2, and TEMP1 (ST_B10, ST_B6) = (DN × 0.00341802 + 149) / 300, kelvin divided by 300 as in the TF records. NDVI and wavelet bands are used as they are. NL is clipped to 0–150 and normalised with `NormFunc` (VIIRS from 2014: mean 0.133, sd 1.748; DMSP: mean 4.153, sd 3.476), as in `26_tfrecords_generation_DMSP_VIIRS`; the TF records had NL raw. The IWI target is not changed, and no statistics of the data are used.
* **PyTorch details.** As in 04, each epoch shuffles the whole training set, the output bias starts at the mean IWI, and training uses bfloat16 autocast on the GPU. The output layer runs in float32.

**Caches and runs.** The bands come from the two caches shared by 04 to 07 (`table1_cache/` and `table2_cache/`, section 3), built by the first run that needs them. The runs go to `table2_runs_fixednorm/`; the earlier runs with standardised bands stay in `table2_runs/`.

**Running.** Check `WORK_DIR` and `TORCH_HOME` in Settings. Each of the 45 runs (9 rows × 5 folds) trains for 100 epochs, and the log prints the time per epoch. Every run saves its state after each epoch, so a stopped job continues where it stopped when you start it again.

**From random weights.** `TRAIN_FROM_SCRATCH=1` in the environment starts VGG-16 from random weights (torchvision's initialisation) and trains all its layers, with learning rate `LR_SCRATCH` = 1e-4: VGG-16 has no BatchNorm, and 1e-3 often stalls. The checkpoints then also hold the VGG-16 layers (about 60 MB for `best.pt` and 180 MB for `last.pt`). Give these runs their own `RUN_ROOT`: a run folder holds one kind of run, and the notebook stops when it finds the other kind.

> **Numbering.** This notebook was written before the notebooks were renumbered for this repository. Apart from the title, its text, comments and plot labels use the earlier numbers: 05 = `04_Table2_DHS`, 07 = `05_Table2_ResNet18`, 08 = `06_NL_ResNet18_Features`, 09 = `07_Landsat_ResNet18_Features`, 10 = `08_Landsat_ViTS16_Features`, 11 = `09_Ridge_NL_Landsat` and 12 = `10_Ridge_NL_ViT`; 01 to 03 are unchanged. The earlier 04 and 06 belonged to the simulated dataset of the preliminary study and are not included. Folder names such as `ridge_runs/nl08_landsat09` keep the earlier numbers.
>
> **Table 2** is Table 2 of the original submission (the DHS experiment of the preliminary study). Its rerun under spatial folds is reported in Supplementary Section S4 of the revised paper.

# 3. Band caches

Every cluster's bands are computed once from its three GeoTIFFs of 01 and stored as float16, in two caches that 04 to 07 share:

* **The band cache** (`table1_cache/`, about 2.5 MB per cluster): the 7 Landsat bands with the GEE scale factors, surface reflectance (DN × 0.0000275 − 0.2) and TEMP1 in kelvin / 300 ((DN × 0.00341802 + 149) / 300); the NDVI series (for `monthly`, each MODIS composite counts for the month in which it starts, and the composites of a month are averaged); and the 6 wavelet bands of that series.
* **The NL cache** (`table2_cache/`, about 0.1 MB per cluster): the `NL.tif` image as it is.

The rest of the fixed normalisation of Settings is applied when the bands are loaded for training (`normalise()`): NL is clipped to 0–150 and normalised with `NormFunc` of the cluster's survey year (and TEMP1 is rescaled if `TEMP1_DIVISOR` is not 300). The table at the end of the next cell shows the resulting training values for a sample of clusters.

A cache whose cluster list differs from the table above is built again. A job that finds a cache being built by another job (of 04 to 07) waits for it.

````python
PATCH = 224
BANDS_DIR = os.path.join(BAND_CACHE, f'{NL_SENSOR}_{NDVI_SERIES}_{WAVELET}_{WAVELET_MODE}')
NL_DIR    = os.path.join(NL_CACHE, f'{NL_SENSOR}_NL')


def month_of_composites(year):
    """Calendar month (1-12) in which each of the 23 MOD13Q1 composites of a year starts (DOY 1, 17, ..., 353)."""
    jan1 = datetime.date(int(year), 1, 1)
    return np.array([(jan1 + datetime.timedelta(days=16 * k)).month for k in range(23)])


def cluster_bands(folder, year):
    """The bands of the band cache for one cluster (float32, CHANNELS order) and its mean NL, or None if a file has
    no-data pixels. TEMP1 is kelvin / 300 here; normalise() applies TEMP1_DIVISOR."""
    with rasterio.open(os.path.join(folder, 'Landsat.tif')) as src:
        ls = src.read().astype(np.float32)
    with rasterio.open(os.path.join(folder, 'MODIS_NDVI.tif')) as src:
        nd = src.read().astype(np.float32)
    with rasterio.open(os.path.join(folder, 'NL.tif')) as src:
        nl = src.read(1).astype(np.float64)
    if (ls == NODATA).any() or (nd == NODATA).any() or (nl == NODATA).any():
        return None
    ls[:6] = ls[:6] * 0.0000275 - 0.2                     # surface reflectance
    ls[6] = (ls[6] * 0.00341802 + 149.0) / 300.0          # kelvin / 300
    if NDVI_SERIES == 'monthly':
        month = month_of_composites(year)
        nd = np.stack([nd[month == m].mean(axis=0) for m in range(1, 13)])
    lfc, hfc = pywt.dwt(nd, WAVELET, mode=WAVELET_MODE, axis=0)    # single-level DWT along time, every pixel
    wav = np.stack([lfc.min(0), lfc.max(0), lfc.mean(0), hfc.min(0), hfc.max(0), hfc.mean(0)])
    return np.concatenate([ls, nd, wav]).astype(np.float32), float(nl.mean())


def cluster_nl(folder, year):
    """The NL band of one cluster (float32, shape (1, 224, 224)) and its mean, or None if it has no-data pixels."""
    with rasterio.open(os.path.join(folder, 'NL.tif')) as src:
        nl = src.read(1).astype(np.float64)
    if (nl == NODATA).any():
        return None
    return nl[None].astype(np.float32), float(nl.mean())


COMPUTE = {'bands': cluster_bands, 'nl': cluster_nl}


def _cache_one(task):
    """Compute one cluster and write it at its place in the .part file. Returns the mean NL and the band sums."""
    i, folder, year, offset, kind, part = task
    try:
        out = COMPUTE[kind](folder, year)
    except Exception as e:
        return i, np.nan, None, f'{type(e).__name__}: {e}'[:200]
    if out is None:
        return i, np.nan, None, 'no-data pixels'
    x, nl_mean = out
    fd = os.open(part, os.O_WRONLY)
    try:
        os.pwrite(fd, x.astype(np.float16).tobytes(), offset)
    finally:
        os.close(fd)
    x64 = x.astype(np.float64)
    return i, nl_mean, np.stack([x64.sum(axis=(1, 2)), (x64 ** 2).sum(axis=(1, 2))]), ''


def build_cache(cache_dir, channels, kind):
    """features.npy, band_stats.csv and index.csv of one cache folder."""
    t0 = time.time()
    features = os.path.join(cache_dir, 'features.npy')
    shape = (len(clusters), len(channels), PATCH, PATCH)
    arr = np.lib.format.open_memmap(features + '.part', mode='w+', dtype=np.float16, shape=shape)
    header, per_cluster = arr.offset, int(np.prod(shape[1:])) * 2
    del arr
    tasks = [(i, r.folder, int(r.year), header + i * per_cluster, kind, features + '.part')
             for i, r in enumerate(clusters.itertuples())]
    nl_mean = np.full(len(clusters), np.nan)
    sums = np.zeros((2, len(channels)))
    error = [''] * len(clusters)
    with multiprocessing.get_context('fork').Pool(N_WORKERS) as pool:
        for i, t, s, err in tqdm(pool.imap_unordered(_cache_one, tasks, chunksize=4), total=len(tasks), desc=kind):
            nl_mean[i], error[i] = t, err
            if s is not None:
                sums += s
    ok = np.isfinite(nl_mean)
    n_pix = ok.sum() * PATCH * PATCH
    mean = sums[0] / n_pix
    std = np.sqrt(np.maximum(sums[1] / n_pix - mean ** 2, 1e-12))
    pd.DataFrame({'band': channels, 'mean': mean, 'std': std}).to_csv(os.path.join(cache_dir, 'band_stats.csv'),
                                                                      index=False)
    os.replace(features + '.part', features)
    index = clusters[['cluster_id', 'country', 'year', 'lat', 'lon']].copy()
    index['nl_mean'], index['ok'], index['error'] = nl_mean, ok, error
    index.to_csv(os.path.join(cache_dir, 'index.csv'), index=False)    # the cache counts as complete once this exists
    print(f'{kind} cache: {ok.sum()} of {len(ok)} clusters in {time.time() - t0:.0f} s -> {features}')


def cache_is_complete(cache_dir):
    index, features = os.path.join(cache_dir, 'index.csv'), os.path.join(cache_dir, 'features.npy')
    if not (os.path.exists(index) and os.path.exists(features)):
        return False
    return pd.read_csv(index)['cluster_id'].tolist() == clusters['cluster_id'].tolist()


def ensure_cache(cache_dir, channels, kind):
    """Build a cache folder unless it is complete. Waits when another job is building it (same lock in 04 to 07)."""
    os.makedirs(cache_dir, exist_ok=True)
    lock = os.path.join(cache_dir, 'building.lock')
    if cache_is_complete(cache_dir):
        return
    try:
        os.mkdir(lock)
    except FileExistsError:
        print(f'another job is building {cache_dir} (if none is running, delete {lock}); waiting ...')
        t_wait = time.time()
        while os.path.exists(lock) and time.time() - t_wait < 4 * 3600:
            time.sleep(30)
        assert cache_is_complete(cache_dir), 'the cache is still not complete: start the notebook again'
        return
    try:
        if not cache_is_complete(cache_dir):           # another job may have finished it just now
            build_cache(cache_dir, channels, kind)
    finally:
        os.rmdir(lock)


def load_cache(cache_dir, channels):
    index = pd.read_csv(os.path.join(cache_dir, 'index.csv'))
    feat = np.load(os.path.join(cache_dir, 'features.npy'), mmap_mode='r')
    assert feat.shape[1:] == (len(channels), PATCH, PATCH), feat.shape
    return index, feat


# ---- the fixed normalisation of the cached bands (Settings) -----------------------------------------------------
TEMP_FROM_CACHE = 300.0 / TEMP1_DIVISOR    # the band cache keeps TEMP1 in kelvin / 300


def nl_norm_params(pos):
    """NormFunc mean and standard deviation (float32 arrays) of the clusters at the cache rows pos."""
    params = [NormFunc(int(y)) for y in clusters['year'].to_numpy()[pos]]
    return np.array([m for m, _ in params], np.float32), np.array([s for _, s in params], np.float32)


def normalise(x, bands, nl_mean, nl_std):
    """Training values of cached bands, in place. x: float32 tensor (clusters, bands, 224, 224); nl_mean, nl_std:
    tensors with one value per cluster (nl_norm_params). Landsat with the GEE scale factors (cached as they are, TEMP1
    in kelvin / TEMP1_DIVISOR), NL clipped to NL_CLIP and normalised with NormFunc, the rest unchanged."""
    for k, b in enumerate(bands):
        if b == 'TEMP1' and TEMP_FROM_CACHE != 1.0:
            x[:, k] *= TEMP_FROM_CACHE
        elif b == 'NL':
            nl = x[:, k] if NL_CLIP is None else x[:, k].clamp(*NL_CLIP)
            x[:, k] = (nl - nl_mean.view(-1, 1, 1)) / nl_std.view(-1, 1, 1)
    return x


ensure_cache(BANDS_DIR, CHANNELS, 'bands')
ensure_cache(NL_DIR, ['NL'], 'nl')
band_index, FEAT = load_cache(BANDS_DIR, CHANNELS)
nl_index, NLF = load_cache(NL_DIR, ['NL'])
complete = band_index['ok'].to_numpy() & nl_index['ok'].to_numpy()
print(f'band cache {FEAT.shape}, {FEAT.nbytes / 1e9:.1f} GB -> {BANDS_DIR}')
print(f'NL cache {NLF.shape}, {NLF.nbytes / 1e9:.1f} GB -> {NL_DIR}')
print(f'clusters with complete data: {complete.sum()} of {len(complete)}; left out:',
      band_index.loc[~complete, 'error'].value_counts().to_dict())

# training values of every band for a sample of clusters
sample = np.sort(np.random.default_rng(0).choice(np.flatnonzero(complete), size=min(32, complete.sum()), replace=False))
m, s = nl_norm_params(sample)
x = torch.from_numpy(np.concatenate([FEAT[sample], NLF[sample]], axis=1).astype(np.float32))
x = normalise(x, CHANNELS + ['NL'], torch.from_numpy(m), torch.from_numpy(s))
print(f'\ntraining values of the input bands, {len(sample)} clusters (NormFunc of',
      sorted({NormFunc(int(y)) for y in clusters['year'].to_numpy()[sample]}), '):')
pd.DataFrame({'min': x.amin(dim=(0, 2, 3)).numpy(), 'mean': x.mean(dim=(0, 2, 3)).numpy(),
              'max': x.amax(dim=(0, 2, 3)).numpy()}, index=CHANNELS + ['NL']).T.round(3)
````

# 4. Model and training

`VGG16Regressor` is `create_model()` of the TF notebooks in PyTorch, as in 04. Only the 1 × 1 convolution and the output layer learn, while the VGG-16 convolution layers keep their ImageNet weights. The output layer runs in float32, because bfloat16 would round IWI values near 50 to steps of 0.25.

With `TRAIN_FROM_SCRATCH` (Settings), VGG-16 starts from torchvision's random initialisation and every layer learns, with `LR_SCRATCH`; `best.pt` and `last.pt` then also hold the VGG-16 layers (`features`). `check_start()` stops the notebook when a run folder holds a run of the other kind (`from_scratch` in `result.json`, `features` in the checkpoints).

`Bands` holds the input bands of one row with the fixed normalisation of Settings (`normalise()`, section 3). They come from the two caches, in the order of the row: the 7 Landsat bands, NL, then the extra bands.

A run trains one row of Table 2 with one test fold. After every epoch it writes the history, the best weights (`best.pt`, lowest validation loss including the L2 term, like Keras `val_loss`) and its full state (`last.pt`). The last step tests `best.pt` on the test fold and writes `test_predictions.csv` and `result.json`. A run with a `result.json` is skipped. A run with only a `last.pt` continues from that epoch.

If loading the ImageNet weights fails because the machine has no internet access, download them once on a machine that has it, then run again:

```bash
     https://download.pytorch.org/models/vgg16-397923af.pth
```

````python
VGG_FEATURES = None
if PRETRAINED:
    try:
        VGG_FEATURES = torchvision.models.vgg16(weights='IMAGENET1K_V1').features.state_dict()
    except Exception as e:
        raise RuntimeError(f'VGG-16 ImageNet weights could not be loaded ({type(e).__name__}: {e}). '
                           'Download them as shown above.') from e


class VGG16Regressor(nn.Module):
    """1x1 conv (bands -> 3, ReLU), frozen VGG-16 convolution layers, flatten, linear -> 1 (create_model()).
    TRAIN_FROM_SCRATCH: the VGG-16 layers start from random weights (torchvision's initialisation) and learn too."""

    def __init__(self, n_bands):
        super().__init__()
        self.adapter = nn.Conv2d(n_bands, 3, kernel_size=1)
        self.features = torchvision.models.vgg16(weights=None).features
        if VGG_FEATURES is not None:
            self.features.load_state_dict(VGG_FEATURES)
        if not TRAIN_FROM_SCRATCH:                        # the ImageNet layers stay fixed
            for p in self.features.parameters():
                p.requires_grad_(False)
        self.head = nn.Linear(512 * 7 * 7, 1)
        for layer in (self.adapter, self.head):           # Keras defaults: Glorot uniform weights, zero bias
            nn.init.xavier_uniform_(layer.weight)
            nn.init.zeros_(layer.bias)

    def forward(self, x):
        x = self.features(F.relu(self.adapter(x)))
        with torch.autocast(device_type=x.device.type, enabled=False):    # output layer in float32
            return self.head(torch.flatten(x, 1).float()).squeeze(1)

    def learned_state(self):
        """What a checkpoint keeps: the 1x1 convolution and the output layer, and with TRAIN_FROM_SCRATCH also the
        VGG-16 layers (features)."""
        state = {'adapter': self.adapter.state_dict(), 'head': self.head.state_dict()}
        if TRAIN_FROM_SCRATCH:
            state['features'] = self.features.state_dict()
        return state

    def load_learned_state(self, state, where):
        check_start('features' in state, where)
        self.adapter.load_state_dict(state['adapter'])
        self.head.load_state_dict(state['head'])
        if TRAIN_FROM_SCRATCH:
            self.features.load_state_dict(state['features'])


def check_start(from_scratch, where):
    """A run folder holds one kind of run: stop instead of mixing runs from pretrained and from random weights
    (TRAIN_FROM_SCRATCH). Runs written before 2026-11-09 have no 'from_scratch' and started from pretrained weights."""
    if bool(from_scratch) != TRAIN_FROM_SCRATCH:
        kind = {True: 'from random weights', False: 'from the pretrained weights'}
        raise RuntimeError(f'{where} holds a run {kind[bool(from_scratch)]}, but this run starts '
                           f'{kind[TRAIN_FROM_SCRATCH]}: give it another RUN_ROOT')


SOURCE = {b: ('bands', CHANNELS.index(b)) for b in CHANNELS}            # where every input band is cached
SOURCE['NL'] = ('nl', 0)


class Bands:
    """The input bands of one row for all clusters, from the two caches, with the fixed normalisation of normalise().
    The cached values go into DEVICE memory when they take less than GPU_DATA_GB, otherwise every batch is read from
    the caches."""

    def __init__(self, bands, rows):
        self.bands = list(bands)
        self.n_bands = len(self.bands)
        self.rows = np.asarray(rows)
        from_bands = [b for b in self.bands if SOURCE[b][0] == 'bands']
        self.ch = np.array([SOURCE[b][1] for b in from_bands], dtype=int)
        self.use_nl = 'NL' in self.bands
        read_order = from_bands + (['NL'] if self.use_nl else [])
        self.order = np.array([read_order.index(b) for b in self.bands])
        nl_m, nl_s = nl_norm_params(self.rows)
        self.nl_mean = torch.from_numpy(nl_m).to(DEVICE)     # NormFunc of every cluster, by position in rows
        self.nl_std = torch.from_numpy(nl_s).to(DEVICE)
        gb = len(self.rows) * self.n_bands * PATCH * PATCH * 2 / 1e9
        self.x = None
        if gb < GPU_DATA_GB:
            self.x = torch.empty((len(self.rows), self.n_bands, PATCH, PATCH), dtype=torch.float16, device=DEVICE)
            for s in tqdm(range(0, len(self.rows), 256), desc=f'loading {self.n_bands} bands, {gb:.1f} GB', leave=False):
                self.x[s:s + 256] = torch.from_numpy(self.read(self.rows[s:s + 256])).to(DEVICE)

    def read(self, pos):
        """float16 cached values of the cache rows pos, in the order of the row."""
        parts = []
        if len(self.ch):
            parts.append(FEAT[pos[:, None], self.ch[None, :]])
        if self.use_nl:
            parts.append(NLF[pos[:, None], np.zeros((1, 1), dtype=int)])
        x = parts[0] if len(parts) == 1 else np.concatenate(parts, axis=1)
        return np.ascontiguousarray(x[:, self.order])

    def batch(self, idx):
        """Normalised float32 bands of the clusters at positions idx (a 1-D tensor)."""
        if self.x is not None:
            x = self.x[idx.to(self.x.device)]
        else:
            x = torch.from_numpy(self.read(self.rows[idx.cpu().numpy()])).to(DEVICE, non_blocking=True)
        idx = idx.to(DEVICE)
        x = normalise(x.float(), self.bands, self.nl_mean[idx], self.nl_std[idx])
        return x.contiguous(memory_format=torch.channels_last)


def r2_score(y, p):
    y, p = np.asarray(y, np.float64), np.asarray(p, np.float64)
    return float(1 - ((y - p) ** 2).sum() / ((y - y.mean()) ** 2).sum())


def autocast():
    return torch.autocast(device_type=DEVICE.type, dtype=torch.bfloat16, enabled=AMP and DEVICE.type == 'cuda')


def predict(model, data, idx, batch=256):
    model.eval()
    out = []
    with torch.no_grad(), autocast():
        for s in range(0, len(idx), batch):
            out.append(model(data.batch(idx[s:s + batch])).float())
    return torch.cat(out)


def save(obj, path):
    torch.save(obj, path + '.tmp')
    os.replace(path + '.tmp', path)
````

````python
def run_fold(key, data, y, folds, test_fold):
    """Train one row of Table 2 with one test fold; returns the test metrics (also in result.json)."""
    run_dir = os.path.join(RUN_ROOT, key, f'test_{test_fold}')
    result_file = os.path.join(run_dir, 'result.json')
    if os.path.exists(result_file):
        with open(result_file) as f:
            result = json.load(f)
        check_start(result.get('from_scratch', False), run_dir)
        return result
    os.makedirs(run_dir, exist_ok=True)
    k = FOLDS.index(test_fold)
    val_fold = FOLDS[(k - 1) % len(FOLDS)]
    pos = lambda mask: torch.from_numpy(np.flatnonzero(mask)).to(DEVICE)
    tr, va, te = pos(~np.isin(folds, [test_fold, val_fold])), pos(folds == val_fold), pos(folds == test_fold)

    torch.manual_seed(SEED + k)
    model = VGG16Regressor(data.n_bands).to(DEVICE).to(memory_format=torch.channels_last)
    with torch.no_grad():
        model.head.bias.fill_(float(y[tr].mean()))
    opt = torch.optim.Adam([p for p in model.parameters() if p.requires_grad], lr=LR, eps=1e-7)   # Keras epsilon
    gen = torch.Generator(device=DEVICE)
    gen.manual_seed(SEED * 100 + k)
    start, best, best_epoch = 0, float('inf'), -1
    last_file, best_file, hist_file = (os.path.join(run_dir, f) for f in ('last.pt', 'best.pt', 'history.csv'))
    if os.path.exists(last_file):                       # continue a run that was stopped
        st = torch.load(last_file, map_location=DEVICE)
        model.load_learned_state(st, run_dir)
        opt.load_state_dict(st['opt'])
        gen.set_state(st['gen'].cpu())
        start, best, best_epoch = st['epoch'] + 1, st['best'], st['best_epoch']
        print(f'{key} test {test_fold}: continuing at epoch {start + 1}')

    for epoch in range(start, EPOCHS):
        t0 = time.time()
        model.train()
        perm = tr[torch.randperm(len(tr), generator=gen, device=DEVICE)]
        sq_sum = torch.zeros((), device=DEVICE)
        n_seen = 0
        for s in range(0, len(perm) - BATCH_SIZE + 1, BATCH_SIZE):     # last incomplete batch dropped, as in TF
            b = perm[s:s + BATCH_SIZE]
            with autocast():
                pred = model(data.batch(b))
            mse = F.mse_loss(pred.float(), y[b])
            loss = mse + L2_HEAD * model.head.weight.pow(2).sum()
            opt.zero_grad(set_to_none=True)
            loss.backward()
            opt.step()
            sq_sum += mse.detach() * len(b)
            n_seen += len(b)
        reg = L2_HEAD * float(model.head.weight.detach().pow(2).sum())
        p_va = predict(model, data, va).cpu().numpy()
        y_va = y[va].cpu().numpy()
        val_mse = float(((p_va - y_va) ** 2).mean())
        val_loss = val_mse + reg
        if val_loss < best:
            best, best_epoch = val_loss, epoch
            save({**model.learned_state(), 'epoch': epoch}, best_file)
        rec = {'epoch': epoch + 1, 'train_mse': float(sq_sum) / max(n_seen, 1), 'l2_term': reg, 'val_mse': val_mse,
               'val_loss': val_loss, 'val_r2': r2_score(y_va, p_va), 'seconds': round(time.time() - t0, 1)}
        pd.DataFrame([rec]).to_csv(hist_file, mode='a', header=not os.path.exists(hist_file), index=False)
        save({**model.learned_state(), 'opt': opt.state_dict(),
              'gen': gen.get_state(), 'epoch': epoch, 'best': best, 'best_epoch': best_epoch}, last_file)
        if epoch == start or (epoch + 1) % 10 == 0 or epoch + 1 == EPOCHS:
            print(f"{key} test {test_fold} epoch {epoch + 1}/{EPOCHS}: train mse {rec['train_mse']:.2f}, "
                  f"val mse {val_mse:.2f}, val R2 {rec['val_r2']:.3f}, best epoch {best_epoch + 1}, {rec['seconds']} s")

    st = torch.load(best_file, map_location=DEVICE)
    model.load_learned_state(st, run_dir)
    p_te = predict(model, data, te).cpu().numpy()
    y_te = y[te].cpu().numpy()
    test_rows = clusters.iloc[data.rows[te.cpu().numpy()]]
    pd.DataFrame({'cluster_id': test_rows['cluster_id'].to_numpy(), 'country': test_rows['country'].to_numpy(),
                  'year': test_rows['year'].to_numpy(), 'y_true': y_te, 'y_pred': p_te}
                 ).to_csv(os.path.join(run_dir, 'test_predictions.csv'), index=False)
    result = {'key': key, 'test_fold': test_fold, 'val_fold': val_fold, 'bands': data.n_bands,
              'n_train': len(tr), 'n_val': len(va), 'n_test': len(te), 'best_epoch': st['epoch'] + 1,
              'test_r2': r2_score(y_te, p_te), 'test_mse': float(((p_te - y_te) ** 2).mean()),
              'test_mae': float(np.abs(p_te - y_te).mean()), 'epochs': EPOCHS, 'from_scratch': TRAIN_FROM_SCRATCH}
    with open(result_file + '.tmp', 'w') as f:
        json.dump(result, f, indent=1)
    os.replace(result_file + '.tmp', result_file)
    print(f"{key} test {test_fold}: test R2 {result['test_r2']:.3f} (best epoch {result['best_epoch']})")
    return result
````

The clusters with complete data and an IWI value. The bands of a row are loaded once and used for its five test folds. Rows and test folds come from `RUN_EXPERIMENTS` and `RUN_FOLDS`.

# 6. Table 2

For every row: the mean and standard deviation of the test R² over the folds that are done, and the R² and RMSE (in IWI points) of all test predictions pooled.
````python
per_fold, pooled = [], {}
for key in TABLE2:
    preds = []
    for f in FOLDS:
        run_dir = os.path.join(RUN_ROOT, key, f'test_{f}')
        if os.path.exists(os.path.join(run_dir, 'result.json')):
            with open(os.path.join(run_dir, 'result.json')) as fh:
                per_fold.append(json.load(fh))
            check_start(per_fold[-1].get('from_scratch', False), run_dir)   # a run folder holds one kind of run
            preds.append(pd.read_csv(os.path.join(run_dir, 'test_predictions.csv')))
    if preds:
        p = pd.concat(preds)
        pooled[key] = (r2_score(p['y_true'], p['y_pred']), float(np.sqrt(((p['y_true'] - p['y_pred']) ** 2).mean())))
per_fold = pd.DataFrame(per_fold, columns=['key', 'test_fold', 'val_fold', 'bands', 'n_train', 'n_val', 'n_test',
                                           'best_epoch', 'test_r2', 'test_mse', 'test_mae', 'epochs'])

rows_out = []
for key, (desc, bands, paper) in TABLE2.items():
    r = per_fold[per_fold['key'] == key]['test_r2']
    r2_pooled, rmse_pooled = pooled.get(key, (np.nan, np.nan))
    rows_out.append({'key': key, 'Table 2 row': desc, 'bands': len(bands), 'paper R2': paper,
                     'Keras fold-5 R2': KERAS_FOLD5[key], 'folds done': len(r),
                     'test R2 mean': r.mean() if len(r) else np.nan, 'test R2 std': r.std() if len(r) > 1 else np.nan,
                     'test R2 min': r.min() if len(r) else np.nan, 'test R2 max': r.max() if len(r) else np.nan,
                     'pooled R2': r2_pooled, 'pooled RMSE': rmse_pooled})
table2 = pd.DataFrame(rows_out)
os.makedirs(RUN_ROOT, exist_ok=True)
for frame, name in ((table2, 'table2_results.csv'), (per_fold, 'table2_per_fold.csv')):
    path = os.path.join(RUN_ROOT, name)
    frame.to_csv(path + '.tmp', index=False)              # parallel runs may write at the same time
    os.replace(path + '.tmp', path)
print(f"{TARGET} target, {NL_SENSOR} clusters, NDVI series '{NDVI_SERIES}', {len(rows)} clusters, spatial folds of 03"
      f"{START_NOTE}\n")
table2.round(3)
````
