# NL experiment · 15 Landsat and the 23 raw MODIS NDVI composites → NL, with a temporal transformer in place of the wavelet bands

This notebook is 11 with the NDVI series fed to the networks as it is. 11 condenses the series of every pixel into 6 wavelet bands: the 23 MODIS composites averaged to 12 months, a single-level `dmey` DWT along time, and the minimum, maximum and mean of its low- and high-frequency parts. These statistics change little when the green season comes earlier or later in the year, and that matters, because the same land use greens up in different months in the Sahel, the Congo basin and southern Africa. Here the 23 composites of the survey year go into the model unchanged (no monthly averaging, no wavelet), and a **temporal transformer** learns the summary instead, end to end with each network. For every cell of 240 m it reads the 23 values as 23 tokens, lets them attend to one another, pools them and returns 6 learned **temporal bands**, which take the place of the 6 wavelet bands of 11. The target is the average nighttime light of the patch (02), and no NL goes in: this is 15 of the IWI study with NL moved from the input to the target. Everything else is as in 11: the 7 Landsat bands, the ten models, the clusters, the spatial folds of 03, the training and the outputs. So every model can be compared with itself in 11 (wavelet bands) and in 12 (Landsat only), test fold by test fold (section 6).

**How the transformer handles the shift** (`TEMPORAL_POSITION`). Self-attention is permutation-equivariant. Without position information it treats its tokens as a set, and pooling then gives a summary that does not change when the composites are reordered; a seasonal shift is such a reordering. But a set also loses the order: one long season and two short ones with the same NDVI values look the same, and so do green-up and senescence. The default keeps the order and still has the invariance:

| `TEMPORAL_POSITION` | what the attention knows about time | series shifted by whole composites | composites reordered |
|---|---|---|---|
| `cyclic` (default) | rotary position embedding (RoPE) with the harmonics 0–7 of the year (23 composites = one period): the attention between two composites depends on their NDVI and on their distance in the year, modulo one year, never on the date | same bands | other bands |
| `none` | nothing: the 23 values are a set (plain self-attention) | same bands | same bands |
| `absolute` | a learned embedding of every composite date, as in a standard transformer | other bands | other bands |

With `cyclic` and `none` the pooling is an attention with learned queries and no position, so the 6 temporal bands of a cell are the same, up to rounding, for its series and for that series shifted by any whole number of composites (16 days each; 23 composites are one year to within 1 %). Section 4 checks this. `absolute` is the date-aware transformer of the satellite time-series literature, the reference that shows what the invariance brings or costs. Each setting writes its own run folder, and section 6 compares them.

**The temporal transformer** (section 4). MOD13Q1 has 250 m pixels, which 01 resampled to the 30 m grid. The transformer reads cells of `NDVI_POOL` × `NDVI_POOL` = 8 × 8 pixels (240 m, the mean of each composite over the cell: 28 × 28 cells per image). Each composite is one token: its NDVI value through a small MLP (width 64). Two pre-norm blocks of self-attention (4 heads) and MLP follow, then attention pooling and a linear map to the 6 temporal bands. Every pixel of a cell gets the bands of its cell. The transformer has about 0.08 M parameters, starts from random weights in every model and learns with the network. Per 30 m pixel (`NDVI_POOL = 1`) it would cost 64 times as much for the same MODIS information.

| key | model | start | how the bands go in | training |
|---|---|---|---|---|

| `unet` | U-Net with an EfficientNet-B3 encoder (torchgeo `unet`) | `Unet_Weights.SENTINEL2_2CLASS_FTW` (Fields of The World: Sentinel-2 B4, B3, B2, B8 at two dates) | RED, GREEN, BLUE and NIR get the weights of their Sentinel-2 band (both dates added), the other 9 bands new weights that start at zero; the decoder gives a 224 × 224 map, and its mean is the predicted NL | `adam` |

| `vgg16bn` | VGG-16 with batch normalisation (torchvision `vgg16_bn`), **baseline** | ImageNet | as `resnet18`: the first convolution takes all 13 bands (the 7 image bands and the 6 temporal bands): RED, GREEN and BLUE get their ImageNet weights, every other band their mean, all × 3/13; global average pooling after the last convolution block gives 512 features | `adam` |
| `mlp` | MLP on band statistics, **baseline** without convolutions | no weights | the mean, standard deviation and 10th, 50th and 90th percentile of every one of the 13 bands after the temporal transformer over the patch (65 values), standardised (BatchNorm), two hidden layers of 256 (ReLU) | `adam` |

The temporal transformer is a new part of every network except `satclip_only`, like the new projections of 11: `adam` trains it with learning rate 1e-3 like every other weight, `adamw` with `LR_NEW` (1e-3, no weight decay, the warm-up and cosine decay of 11). The networks whose new projection starts at zero (Swin, U-Net, ViT, OlmoEarth) start as the pretrained model and learn to use the temporal bands, as they learned to use the wavelet bands in 11.

The NDVI composites are read as 01 wrote them (NDVI, the MODIS values × 0.0001), and the transformer reads (NDVI − 0.5) / 0.25: fixed numbers, like the rest.

**Baselines.** `vgg16bn` and `mlp` are trained on exactly the same inputs, clusters, folds and recipe as the ten models, from ImageNet weights and from scratch (`TRAIN_FROM_SCRATCH=1`; the MLP has no pretrained weights, so its two runs differ only by training noise). VGG-16 has batch normalisation, so that `adam` with learning rate 1e-3 trains it from either start. The MLP sees only per-band statistics of the patch, not where things are, so its gap to the CNNs measures what spatial patterns add. Their runs go to `RUN_ROOT/vgg16bn/` and `RUN_ROOT/mlp/`, where 16–18 find them.

**SatCLIP.** torchgeo's SatCLIP is the location encoder of SatCLIP: it turns a longitude and latitude into 256 numbers. It does not read an image, so it cannot take the 13 bands. `satclip` adds it to ResNet-18 on the 13 bands, and `satclip_only` uses it alone (it reads no band, so it is the same model in 11–15: with the same clusters, folds and seed it gives the same R² up to the nondeterminism of the GPU). The folds of 03 put every country into every fold, so `satclip_only` shows how much of the nighttime light the region alone explains, and `satclip` against `resnet18` what the location adds to the image. The SatCLIP values are an input, standardised with the training clusters of each model. The DHS displacement of the coordinates (up to 2 km for urban, 5 km for rural and 10 km for 1 % of the rural clusters) is small against the scale that SatCLIP L40 resolves (hundreds of km); but the NL target is the light of the patch around the displaced position, so it is consistent with the Landsat and MODIS patches.

**The Landsat bands of Swin and OlmoEarth.** Both were pretrained on Landsat 8/9 Level-1 data, bands B1–B11, as digital numbers (TOA reflectance = DN × 2e-5 − 0.1). 01 downloads surface reflectance and surface temperature (Collection 2 Level 2) of Landsat 5, 7, 8 and 9. The notebook takes the reflectance as if it were TOA reflectance, DN = (reflectance + 0.1) / 2e-5, and fills the bands that are not in the 01 files: B1 (coastal) = BLUE, B8 (panchromatic) = the mean of GREEN and RED, B9 (cirrus) = 5000 (clear sky, about 0), and B10 and B11 (thermal) from TEMP1 with the Planck constants of Landsat 8 TIRS. Swin then gets the normalisation of SatlasPretrain, clip((DN − 4000) / 16320, 0, 1), and OlmoEarth its own (mean ± 2 sd of its pretraining data, from `olmoearth_pretrain_minimal`). For OlmoEarth the B8 tokens are left out by default (`OLMOEARTH_PAN`). These are approximations, which fine-tuning has to absorb.

**Input normalisation**, before the mapping of each model: fixed numbers, as in the IWI study. Landsat with the GEE scale factors (surface reflectance; TEMP1 in kelvin / 300); the MODIS bands as they are. ViT gets the SR bands as in 08 (clipped to 0–0.3 and divided by 0.3), U-Net RED, GREEN, BLUE and NIR divided by 0.3 (FTW: Sentinel-2 DN / 3000), and ResNet the values as they are. **No nighttime light goes in**: it is the target.

**The target** (`TARGET = 'nl'`, a column of `train_clusters_folds_nl.csv`, written by 02 and 03): the mean over the 224 × 224 pixels of the `NL.tif` of the survey year, clipped to 0–150 and normalised with `NormFunc` of the survey year, which is how the NL input band of the IWI study was normalised (DMSP up to 2013: (NL − 4.153) / 3.476; VIIRS from 2014: (NL − 0.133) / 1.748). Within a period this is linear in the patch mean, so the R² of each period (`pooled R2 DMSP`, `pooled R2 VIIRS` in section 6) is the R² in sensor units; the pooled R² of both periods also counts the different levels of the two scales (a dark DMSP patch is −1.195, a dark VIIRS patch −0.076). A prediction times `nl_norm_sd` plus `nl_norm_mean` (columns of the folds file) is the patch mean in sensor units.

**Clusters and folds.** The clusters and the five spatial folds of 03 of the IWI study (`train_clusters_folds_nl.csv` lists exactly its clusters, in its order, and the notebook checks this against `IWI_FOLDS_CSV`): by default all clusters of both periods (DMSP up to 2013, VIIRS from 2014); `NL_SENSORS = ['VIIRS']` (or `['DMSP']`) trains on one period only. Clusters with a no-data pixel in any of their three files are left out, as in the IWI study, so the training, validation and test clusters of every fold are those of the same notebook there. Every model is trained five times: test fold f, validation fold f − 1, training on the other three.

| model | test fold | validation fold | training folds | Keras name |
|---|---|---|---|---|
| `test_E` | E | D | A, B, C | Fold12345 |
| `test_A` | A | E | B, C, D | Fold23451 |
| `test_B` | B | A | C, D, E | Fold34512 |
| `test_C` | C | B | D, E, A | Fold45123 |
| `test_D` | D | C | E, A, B | Fold51234 |

**Training.** MSE loss on the NL target, batch 64, 100 epochs, no augmentation, and the weights with the lowest validation loss, as in the IWI study. The output layer runs in float32, and its bias starts at the mean target of the training clusters. `adam` (the CNNs and SatCLIP): Adam with learning rate 1e-3 for all weights. `adamw` (the transformers): AdamW with 5e-5 for the pretrained weights and 1e-3 for the new ones, weight decay 0.05, 5 epochs of warm-up and cosine decay, stochastic depth 0.1.

**From random weights.** `TRAIN_FROM_SCRATCH=1` in the environment starts every network from random weights (runs in `<periods>_scratch_<position>`); the transformers' weights then learn with `LR_SCRATCH` (5e-5). The SatCLIP values stay those of the pretrained encoder (an input, not a trained part), and MOSAIKS has no pretrained weights. The temporal transformer starts from random weights in both cases.

**Output,** in `WaveletNL/landsat_temporal_runs/<periods>_<start>_<position>/<model>/test_<fold>/` (by default `DMSP_VIIRS_pretrained_cyclic/`):

* `result.json` (with the settings of the temporal transformer), `test_predictions.csv`, and `predictions.csv`: the target predicted for every cluster with complete data, with its `role` (train, val or test) in that model (columns `nl` and `nl_pred`).
* `history.csv` and `best.pt` (the weights, the temporal transformer included) of the networks; `mosaiks.pt` (filters, band scaling and ridge) for MOSAIKS.
* `features.csv` (`SAVE_FEATURES`): the layer before the output for every cluster, columns `lt_<model>_feat_0`, … (11: `lw_`, 12: `ls_`), so that the files can be merged on `cluster_id` (not for MOSAIKS).
* Section 6: `lt_results.csv`, `lt_per_fold.csv`, `oof_predictions.csv` (every cluster once, with the prediction of each model whose test fold it is in), `lt_r2.png`, `lt_scatter.png` and `lt_temporal_bands.png` (the learned temporal bands of one cluster next to its wavelet bands of 11).

**Running.** Run 02 and 03 first: they write `train_clusters_folds_nl.csv`. Check `DATA_DIR`, `TORCH_HOME` and `PY_PACKAGES` in Settings. The weights and `olmoearth_pretrain_minimal` are those of the IWI study (section 4). The band cache and the NDVI cache of 15 of the IWI study (the 23 composites per 240 m cell, about 36 kB per cluster) are used as they are, and a missing one is built first. `TEMPORAL_POSITION=none` or `absolute` in the environment runs the other position settings. Every network saves its state after each epoch, so a stopped job continues where it stopped. For the comparison of section 6, run 11 and 12 too (with the same `NL_SENSORS` and start).

````python
# ---- input / output ------------------------------------------------------------------------
DATA_DIR      = '/path/to/data/'   # the IWI study: Dataset/, its
                                                                    # folds (03) and the caches of its 04-15
WORK_DIR      = os.path.join(DATA_DIR, 'WaveletNL')                 # this study: the folds file with the target, runs
FOLDS_CSV     = os.path.join(WORK_DIR, 'train_clusters_folds_nl.csv')   # 03_Folds_NL_Target: the folds + the target
IWI_FOLDS_CSV = os.path.join(DATA_DIR, 'train_clusters_folds.csv')  # 03 of the IWI study: FOLDS_CSV lists its clusters
BAND_CACHE    = os.path.join(DATA_DIR, 'table1_cache')              # Landsat, NDVI, wavelet bands, ~2.5 MB per cluster,
                                                                    # and the NDVI cache of 15 (~36 kB)
TORCH_HOME    = '/path/to/torch_home'   # pretrained weights: hub/checkpoints
PY_PACKAGES   = '/path/to/python_packages'   # olmoearth_pretrain_minimal,
                                                                    # installed there with pip --target (section 4)

# ---- data (as in 04-15 of the IWI study, so that its caches are shared) -------------------------------------
TARGET       = 'nl'                 # the NL target of 02: the mean of the NL image of the patch, clipped to NL_CLIP
                                    # and normalised with NormFunc of the survey year (a column of FOLDS_CSV)
NL_SENSORS   = ['DMSP', 'VIIRS']    # the periods of 01 (the sensor of the target): DMSP = surveys up to 2013, VIIRS
                                    # = from 2014. Both: all clusters; ['VIIRS']: the clusters of 04 and 05
NDVI_SERIES  = 'monthly'            # of the band cache (only its Landsat bands are read here: the NDVI series of
WAVELET      = 'dmey'               # 15 is the 23 composites, in their own cache); the band cache is built with
WAVELET_MODE = 'symmetric'          # these if it is missing, so that 04-15 can use it
NODATA       = -999999.0            # no-data value of the 01 files; clusters with any no-data pixel are left out

# ---- normalisation of the inputs (fixed numbers, as in 04-15 of the IWI study) ---------------------------------
# Landsat: the Collection 2 Level 2 scale factors of the GEE catalog (LANDSAT/LC08/C02/T1_L2; the same for Landsat 5,
# 7 and 9): SR_B* (BLUE ... SWIR2) DN * 0.0000275 - 0.2 = surface reflectance; ST_B10 / ST_B6 (TEMP1)
# DN * 0.00341802 + 149.0 = surface temperature in kelvin, divided by TEMP1_DIVISOR. NDVI: the composites as 01 wrote
# them (MODIS x 0.0001); the temporal transformer reads (NDVI - NDVI_CENTER) / NDVI_SCALE. Each model then maps the
# image bands to what its pretrained weights expect (section 4). The nighttime light is not an input: it is the
# target. Nighttime light: clipped to NL_CLIP, then (NL - Mean) / stdDev with NormFunc of the survey year, averaged
# over the patch (02). NL_CLIP and NormFunc are those of 02; here they only check the target (section 3).
TEMP1_DIVISOR = 300.0      # kelvin / 300 as in the paper's Keras notebooks
NL_CLIP = (0, 150)         # as in 26_tfrecords_generation_DMSP_VIIRS (VIIRS max 150, DMSP max 63); None: no clipping
NDVI_CENTER = 0.5          # the transformer reads (NDVI - 0.5) / 0.25: NDVI -0.2 ... 1 becomes -2.8 ... 2
NDVI_SCALE  = 0.25


def NormFunc(Year):
    if (Year >= 2014):     # VIIRS
        Mean = 0.1326654328849849
        stdDev = 1.7479698750873318
    else:                  # DMSP
        Mean = 4.152879625271373
        stdDev = 3.4762732210548317

    return Mean, stdDev


# ---- the temporal transformer (section 4) -----------------------------------------------------------------------
N_COMPOSITES      = 23     # MOD13Q1 16-day composites of the survey year (DOY 1, 17, ..., 353): one token each
NDVI_POOL         = 8      # the transformer reads cells of 8 x 8 pixels (240 m, about one 250 m MODIS pixel), the mean
                           # of each composite over the cell: 28 x 28 cells per image (1: every 30 m pixel, 64 x the
                           # work)
POSITIONS = ('cyclic', 'none', 'absolute')
TEMPORAL_POSITION = os.environ.get('TEMPORAL_POSITION', 'cyclic').strip()   # what the attention knows about time:
                           # 'cyclic' = RoPE with the harmonics of the year (distance in the year modulo one year:
                           # invariant to seasonal shifts, order kept), 'none' = nothing (a set of 23 values: invariant
                           # to any reordering), 'absolute' = a learned embedding of every date (not invariant)
TEMPORAL_DIM      = 64     # width of the tokens
TEMPORAL_HEADS    = 4      # attention heads, 16 channels each (cyclic: 8 rotary pairs, the harmonics 0-7 of the year)
TEMPORAL_LAYERS   = 2      # pre-norm transformer blocks (self-attention, then MLP)
TEMPORAL_MLP      = 2      # MLP width = TEMPORAL_MLP x TEMPORAL_DIM
TEMPORAL_DROPOUT  = 0.0    # dropout on the attention weights and in the MLP
TEMPORAL_BANDS    = 6      # learned bands per cell, in place of the 6 wavelet bands of 11
TEMPORAL_CHUNK    = 32768  # cells per pass through the transformer (a batch of 64 images has 50 176 cells; much
                           # larger passes can exceed the grid limits of CUDA's attention kernels)

# ---- training (section 4) -------------------------------------------------------------------------------------
FOLDS         = ['A', 'B', 'C', 'D', 'E']
EPOCHS        = 100
BATCH_SIZE    = 64
L2_HEAD       = 0.0        # weight penalty on the output layer: none, as in 07, 08 and 11 (04, 05: 0.01, the paper)
LR            = 1e-3       # 'adam' (ResNets, U-Net, SatCLIP): Adam, one learning rate, no schedule, as in 05-07 and 11
LR_PRETRAINED = 5e-5       # 'adamw' (ViT, Swin, OlmoEarth), as in 08 and 11: AdamW learning rate of the pretrained
LR_NEW        = 1e-3       # weights, and of the new ones (the output layer, the projection of the extra bands and the
                           # temporal transformer)
LR_SCRATCH    = 5e-5       # LR_PRETRAINED with TRAIN_FROM_SCRATCH (08; DeiT from scratch: 5e-4 x 64 / 512 = 6e-5)
WEIGHT_DECAY  = 0.05       # 'adamw': on the pretrained weight matrices (not biases, norms, embeddings, position biases)
WARMUP_EPOCHS = 5          # 'adamw': linear warm-up of the learning rates, then cosine decay to 0 at EPOCHS
DROP_PATH     = 0.1        # stochastic depth of the transformer blocks (ViT, Swin, OlmoEarth)
SEED          = 3127
AMP           = True       # bfloat16 autocast on the GPU for training; features and predictions in float32
GPU_DATA_GB   = 60         # the 7 image bands go into GPU memory when they need less than this (40 000 clusters: 28 GB;
                           # the 23 composites per cell, 1.4 GB, always do), otherwise every batch is read from the
                           # caches (slower). Lower it if a model runs out of GPU memory.
N_WORKERS     = len(os.sched_getaffinity(0))   # processes that build a cache if it is missing

# ---- model details (section 4, as in 11) -------------------------------------------------------------------------
SR_MAX           = 0.3          # ViT (SSL4EO-L): reflectance clipped to 0-SR_MAX and divided by SR_MAX, as in 08
FTW_SCALE        = 0.3          # U-Net (FTW): Sentinel-2 DN / 3000 = reflectance / 0.3
SWIN_SIZE        = 256          # Swin: the image is resized to this, so that every window of 8 x 8 tokens is full
OLMOEARTH_SIZE   = 'BASE'       # OlmoEarthV1_Weights: NANO, TINY, BASE or LARGE
OLMOEARTH_PATCH  = 8            # OlmoEarth tokens of 8 x 8 pixels of 30 m: 28 x 28 tokens per image
OLMOEARTH_PAN    = 'missing'    # Landsat B8 (panchromatic) is not in the 01 files: 'missing' leaves its tokens out;
                                # 'copy' fills it with the mean of GREEN and RED (twice the tokens, slower)
MOSAIKS_FEATURES = 4096         # MOSAIKS of torchgeo: 4096 features (2048 filters), kernel 4, bias -1
MOSAIKS_KERNEL   = 4
MOSAIKS_BIAS     = -1.0
MOSAIKS_BATCH    = 8            # images per batch for the features: the convolution output is 0.4 GB per image
ALPHAS           = np.logspace(-2, 6, 33)   # MOSAIKS ridge: the alpha with the lowest validation MSE (as 09 of the IWI study)
MLP_QUANTILES    = (0.1, 0.5, 0.9)   # mlp (baseline): percentiles of every band over the patch, besides its mean
                                     # and standard deviation
MLP_HIDDEN       = (256, 256)        # mlp: hidden layers (ReLU) after the standardised band statistics
SAVE_FEATURES    = True         # features.csv of every network: the layer before the output for every cluster

# ---- what to run: models ('all', 'none' = only the caches and the results, or keys of MODELS) and test folds ('all'
# or folds), comma-separated. These can also be set in the environment.
RUN_MODELS = os.environ.get('LT_MODELS', 'all')
RUN_FOLDS  = os.environ.get('LT_FOLDS', 'all')
RUN_FOLDS  = FOLDS if RUN_FOLDS == 'all' else [f.strip() for f in RUN_FOLDS.split(',')]
assert set(RUN_FOLDS) <= set(FOLDS), f'unknown test folds: {RUN_FOLDS}'
assert set(NL_SENSORS) <= {'DMSP', 'VIIRS'}, NL_SENSORS
assert OLMOEARTH_PAN in ('missing', 'copy'), OLMOEARTH_PAN
assert TEMPORAL_POSITION in POSITIONS, f'TEMPORAL_POSITION must be one of {POSITIONS}, not {TEMPORAL_POSITION!r}'
assert TEMPORAL_DIM % TEMPORAL_HEADS == 0 and (TEMPORAL_DIM // TEMPORAL_HEADS) % 2 == 0, \
    'TEMPORAL_DIM / TEMPORAL_HEADS must be an even number (RoPE turns pairs of channels)'
assert 224 % NDVI_POOL == 0, 'NDVI_POOL must divide 224'
KERAS_NAME = {'E': 'Fold12345', 'A': 'Fold23451', 'B': 'Fold34512', 'C': 'Fold45123', 'D': 'Fold51234'}  # 27_Train_NL

# ---- TRAIN_FROM_SCRATCH=1 in the environment: every network starts from random
# weights, the transformers with LR_SCRATCH; the runs go to <periods>_scratch_<position>. check_start() stops the
# notebook when a run folder holds a run of the other kind, or one with another temporal transformer.
_scratch = os.environ.get('TRAIN_FROM_SCRATCH', '0').strip()
assert _scratch in ('', '0', '1'), f'TRAIN_FROM_SCRATCH must be 0 or 1, not {_scratch!r}'
TRAIN_FROM_SCRATCH = _scratch == '1'
PRETRAINED = not TRAIN_FROM_SCRATCH
if TRAIN_FROM_SCRATCH:
    LR_PRETRAINED = LR_SCRATCH
START = 'scratch' if TRAIN_FROM_SCRATCH else 'pretrained'

# the settings that make a run of 15: written into result.json and the checkpoints, and checked when a run continues
TEMPORAL_CONFIG = {'position': TEMPORAL_POSITION, 'ndvi_pool': NDVI_POOL, 'composites': N_COMPOSITES,
                   'bands': TEMPORAL_BANDS, 'dim': TEMPORAL_DIM, 'heads': TEMPORAL_HEADS, 'layers': TEMPORAL_LAYERS,
                   'mlp': TEMPORAL_MLP, 'dropout': TEMPORAL_DROPOUT, 'ndvi_center': NDVI_CENTER,
                   'ndvi_scale': NDVI_SCALE}

# ---- output -------------------------------------------------------------------------------------------------
PERIODS  = '_'.join(NL_SENSORS)
RUN_ROOT = os.path.join(WORK_DIR, 'landsat_temporal_runs', f'{PERIODS}_{START}_{TEMPORAL_POSITION}')     # per model
# for the comparison (section 6): the same models with the 6 wavelet bands (11) and with Landsat only (12), same periods
# and start; the runs of 15 with the other position settings; and as in 11: 07 (landsat_wavelet6) on the same clusters,
# and 05 (base_wavelet6 with ResNet-18, VIIRS clusters only)
WAVELET_RUNS   = os.path.join(WORK_DIR, 'landsat_wavelet_runs', f'{PERIODS}_{START}')        # 11
LANDSAT_RUNS   = os.path.join(WORK_DIR, 'landsat_runs', f'{PERIODS}_{START}')                # 12
POSITION_RUNS  = {p: os.path.join(WORK_DIR, 'landsat_temporal_runs', f'{PERIODS}_{START}_{p}')
                  for p in POSITIONS if p != TEMPORAL_POSITION}                               # 15, other settings
REF_STARTS    = ['scratch', 'imagenet'] if TRAIN_FROM_SCRATCH else ['imagenet', 'scratch']   # first one found is used
T2_RESNET18   = os.path.join(WORK_DIR, 'table2_resnet18_runs_fixednorm', 'finetune')

os.environ['TORCH_HOME'] = TORCH_HOME
if os.path.isdir(PY_PACKAGES) and PY_PACKAGES not in sys.path:
    sys.path.append(PY_PACKAGES)
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
torch.backends.cudnn.benchmark = True
torch.backends.cuda.matmul.allow_tf32 = True
torch.backends.cudnn.allow_tf32 = True
print('device:', DEVICE, torch.cuda.get_device_name(0) if DEVICE.type == 'cuda' else '(no GPU: only for code tests)')
print('torch', torch.__version__, '| torchvision', torchvision.__version__, '| pywt', pywt.__version__,
      '| cache workers:', N_WORKERS)
print('periods:', NL_SENSORS, '| target:', TARGET, '| start:', START, '| temporal transformer:', TEMPORAL_POSITION, 'position,',
      f'{TEMPORAL_LAYERS} blocks of width {TEMPORAL_DIM}, {TEMPORAL_BANDS} bands per cell of {NDVI_POOL} x {NDVI_POOL}',
      'pixels | runs ->', RUN_ROOT)
````

# 1. Input bands and models

The networks see 13 bands, in this order: the 7 Landsat bands and the 6 temporal bands `TT_1` … `TT_6` that the temporal transformer makes of the 23 composites `NDVI_01` … `NDVI_23` (MOD13Q1, starting on DOY 1, 17, …, 353 of the survey year). No nighttime light: it is the target. The data hold the 7 image bands (`IMAGE`) and the composites; the temporal bands are computed inside every network, batch by batch, because they learn. `EXTRA` are the bands that the Landsat weights of ViT, Swin and OlmoEarth do not have (TEMP1 and the temporal bands); they go through the new projection of those models. MOSAIKS reads the 7 image bands and the 23 composites (`MOSAIKS_INPUT`).

````python
LANDSAT = ['BLUE', 'GREEN', 'RED', 'NIR', 'SWIR1', 'SWIR2', 'TEMP1']
N_NDVI = 12 if NDVI_SERIES == 'monthly' else 23
NDVI = [f'NDVI_{k:02d}' for k in range(1, N_NDVI + 1)]           # NDVI bands of the band cache (not read)
WAVELET_BANDS = ['LFC_min', 'LFC_max', 'LFC_mean', 'HFC_min', 'HFC_max', 'HFC_mean']
CHANNELS = LANDSAT + NDVI + WAVELET_BANDS         # the bands of the band cache (04-15), in this order
COMPOSITES = [f'NDVI_{k:02d}' for k in range(1, N_COMPOSITES + 1)]   # the bands of MODIS_NDVI.tif (the cache of 15)
COMPOSITE_DOY = [1 + 16 * k for k in range(N_COMPOSITES)]           # the day of the year each composite starts on
TEMPORAL = [f'TT_{k}' for k in range(1, TEMPORAL_BANDS + 1)]        # the learned temporal bands
IMAGE = LANDSAT                                    # the 7 image bands of the band cache (no NL: the target)
INPUT = IMAGE + TEMPORAL                           # the 13 bands the networks see, in this order
BAND = {b: k for k, b in enumerate(INPUT)}         # position of every band in the input
SR_BANDS = LANDSAT[:6]                             # surface reflectance, BLUE ... SWIR2
EXTRA = ['TEMP1'] + TEMPORAL                       # not in the Landsat weights of ViT, Swin and OlmoEarth
EXTRA_INDEX = [BAND[b] for b in EXTRA]
MOSAIKS_INPUT = IMAGE + COMPOSITES                 # MOSAIKS: 30 bands, no temporal transformer

# key: (model, pretrained weights, training recipe, inputs: 'image' = the 7 image bands, 'ndvi' = the 23 composites
# per cell, 'location' = SatCLIP values)
MODELS = {
    'resnet18':     ('ResNet-18', 'ImageNet (torchvision)', 'adam', ('image', 'ndvi')),
    'resnet34':     ('ResNet-34', 'ImageNet (torchvision)', 'adam', ('image', 'ndvi')),
    'resnet50':     ('ResNet-50', 'ImageNet (torchvision)', 'adam', ('image', 'ndvi')),
    'swin':         ('Swin Transformer v2-B', 'Swin_V2_B_Weights.LANDSAT_SI_SATLAS', 'adamw', ('image', 'ndvi')),
    'unet':         ('U-Net, EfficientNet-B3 encoder', 'Unet_Weights.SENTINEL2_2CLASS_FTW', 'adam', ('image', 'ndvi')),
    'vit':          ('ViT-S/16', 'ViTSmall16_Weights.LANDSAT_ETM_SR_MOCO', 'adamw', ('image', 'ndvi')),
    'satclip':      ('ResNet-18 + SatCLIP location', 'ImageNet + SatCLIP_Weights.SATCLIP_VIT16_L40', 'adam',
                     ('image', 'ndvi', 'location')),
    'satclip_only': ('SatCLIP location + MLP', 'SatCLIP_Weights.SATCLIP_VIT16_L40', 'adam', ('location',)),
    'mosaiks':      ('MOSAIKS (RCF) + ridge', 'none (random convolutional features)', 'ridge', ('image', 'ndvi')),
    'olmoearth':    (f'OlmoEarth v1 {OLMOEARTH_SIZE.title()}', f'OlmoEarthV1_Weights.{OLMOEARTH_SIZE}', 'adamw',
                     ('image', 'ndvi')),
    # baselines: a VGG-16 and an MLP without convolutions, on the same inputs
    'vgg16bn':      ('VGG-16 with batch normalisation', 'ImageNet (torchvision vgg16_bn)', 'adam', ('image', 'ndvi')),
    'mlp':          ('MLP on band statistics', 'none (random weights)', 'adam', ('image', 'ndvi')),
}


def ndvi_route(key):
    """How the NDVI series goes into a model."""
    if key == 'mosaiks':
        return '23 composites as bands (no transformer)'
    if 'ndvi' in MODELS[key][3]:
        return f'temporal transformer ({TEMPORAL_POSITION}) -> {TEMPORAL_BANDS} bands'
    return 'none'


if RUN_MODELS == 'all':
    todo = list(MODELS)
elif RUN_MODELS == 'none':
    todo = []
else:
    todo = [k.strip() for k in RUN_MODELS.split(',') if k.strip()]
bad = [k for k in todo if k not in MODELS]
assert not bad, f'unknown models: {bad}'
print(f'{len(INPUT)} input bands of the networks:', ', '.join(INPUT))
print(f'the temporal bands come from the {N_COMPOSITES} composites (DOY {COMPOSITE_DOY[0]}, {COMPOSITE_DOY[1]}, ..., '
      f'{COMPOSITE_DOY[-1]}) of every {NDVI_POOL} x {NDVI_POOL}-pixel cell')
print('models to run:', todo, '| test folds:', RUN_FOLDS)


def start_of(key):
    """The pretrained weights a model starts from in this run (TRAIN_FROM_SCRATCH: random, except SatCLIP). The
    temporal transformer always starts from random weights."""
    if PRETRAINED or MODELS[key][2] == 'ridge':
        return MODELS[key][1]
    return 'random + SatCLIP_VIT16_L40 (fixed)' if 'location' in MODELS[key][3] else 'random'


pd.DataFrame([(k, m, start_of(k), r, ' + '.join(n), ndvi_route(k)) for k, (m, _, r, n) in MODELS.items()],
             columns=['key', 'model', 'start', 'training', 'input', 'NDVI series'])
````

# 3. Caches

The bands come from two caches of the IWI study (`DATA_DIR`), one folder per period, both float16:

* **The band cache** of 04 to 15 (`table1_cache/<sensor>_monthly_dmey_symmetric/`, about 2.5 MB per cluster): the 7 Landsat bands with the GEE scale factors, surface reflectance (DN × 0.0000275 − 0.2) and TEMP1 in kelvin / 300 ((DN × 0.00341802 + 149) / 300); the 12 monthly NDVI values and the 6 wavelet bands of 11. This notebook reads only its 7 Landsat bands.
* **The NDVI cache** of 15 of the IWI study (`table1_cache/<sensor>_composites23_pool8/`, about 36 kB per cluster): the 23 composites of `MODIS_NDVI.tif` as 01 wrote them (NDVI), each averaged over the cells of `NDVI_POOL` × `NDVI_POOL` pixels, so 23 × 28 × 28 values. 01 resampled the 250 m MODIS pixels (231.7 m in the sinusoidal grid) to the 30 m grid with nearest neighbour, so a cell of 8 × 8 pixels holds about one MODIS pixel; where a cell meets two of them, the mean mixes them a little.

A cluster with a no-data pixel in any of its three files is left out, as in the IWI study (the band cache checks the NL file too), so the clusters are those of 11 (this also leaves out the year-2000 clusters, whose first three composites do not exist). A complete cache is used as it is, a missing one is built, and a job that finds a cache being built by another job waits for it. The NL image is not read for training: the target comes from `FOLDS_CSV`. The rest of the fixed normalisation is applied when the bands are loaded (`normalise()`: TEMP1 rescaled if `TEMP1_DIVISOR` is not 300). The composites stay NDVI; the temporal transformer maps them itself. The table at the end shows the resulting values of every band and composite for a sample of clusters.

````python
PATCH = 224
CELLS = PATCH // NDVI_POOL                         # cells per side of the NDVI grid (28 with NDVI_POOL = 8)
BAND_DIRS = {s: os.path.join(BAND_CACHE, f'{s}_{NDVI_SERIES}_{WAVELET}_{WAVELET_MODE}') for s in NL_SENSORS}
NDVI_DIRS = {s: os.path.join(BAND_CACHE, f'{s}_composites{N_COMPOSITES}_pool{NDVI_POOL}') for s in NL_SENSORS}


def month_of_composites(year):
    """Calendar month (1-12) in which each of the 23 MOD13Q1 composites of a year starts (DOY 1, 17, ..., 353)."""
    jan1 = datetime.date(int(year), 1, 1)
    return np.array([(jan1 + datetime.timedelta(days=16 * k)).month for k in range(23)])


def cluster_bands(folder, year):
    """The bands of the band cache for one cluster (float32, CHANNELS order) and its mean NL, or None if a file has
    no-data pixels. TEMP1 is kelvin / 300 here; normalise() applies TEMP1_DIVISOR. (As in 04-12: this cache is
    shared.)"""
    with rasterio.open(os.path.join(folder, 'Landsat.tif')) as src:
        ls = src.read().astype(np.float32)
    with rasterio.open(os.path.join(folder, 'MODIS_NDVI.tif')) as src:
        nd = src.read().astype(np.float32)
    with rasterio.open(os.path.join(folder, 'NL.tif')) as src:
        nl = src.read(1).astype(np.float64)
    if (ls == NODATA).any() or (nd == NODATA).any() or (nl == NODATA).any():
        return None
    ls[:6] = ls[:6] * 0.0000275 - 0.2                     # surface reflectance
    ls[6] = (ls[6] * 0.00341802 + 149.0) / 300.0          # kelvin / 300
    if NDVI_SERIES == 'monthly':
        month = month_of_composites(year)
        nd = np.stack([nd[month == m].mean(axis=0) for m in range(1, 13)])
    lfc, hfc = pywt.dwt(nd, WAVELET, mode=WAVELET_MODE, axis=0)    # single-level DWT along time, every pixel
    wav = np.stack([lfc.min(0), lfc.max(0), lfc.mean(0), hfc.min(0), hfc.max(0), hfc.mean(0)])
    return np.concatenate([ls, nd, wav]).astype(np.float32), float(nl.mean())


def cluster_nl(folder, year):
    """The NL band of one cluster (float32, shape (1, 224, 224)) and its mean, or None if it has no-data pixels."""
    with rasterio.open(os.path.join(folder, 'NL.tif')) as src:
        nl = src.read(1).astype(np.float64)
    if (nl == NODATA).any():
        return None
    return nl[None].astype(np.float32), float(nl.mean())


def cluster_composites(folder, year):
    """The 23 NDVI composites of one cluster, each averaged over cells of NDVI_POOL x NDVI_POOL pixels (float32, shape
    (23, CELLS, CELLS)), and their mean, or None if MODIS_NDVI.tif has no-data pixels (a missing composite included).
    The other two files are checked by the band cache."""
    with rasterio.open(os.path.join(folder, 'MODIS_NDVI.tif')) as src:
        nd = src.read().astype(np.float64)
    if nd.shape != (N_COMPOSITES, PATCH, PATCH):
        raise ValueError(f'MODIS_NDVI.tif has shape {nd.shape}, not {(N_COMPOSITES, PATCH, PATCH)}')
    if (nd == NODATA).any():
        return None
    cells = nd.reshape(N_COMPOSITES, CELLS, NDVI_POOL, CELLS, NDVI_POOL).mean(axis=(2, 4))
    return cells.astype(np.float32), float(cells.mean())


COMPUTE = {'bands': cluster_bands, 'nl': cluster_nl, 'composites': cluster_composites}
SUMMARY = {'bands': 'nl_mean', 'nl': 'nl_mean', 'composites': 'ndvi_mean'}   # the column of index.csv (04-15: nl_mean)


def _cache_one(task):
    """Compute one cluster and write it at its place in the .part file. Returns its summary value (the mean NL or
    NDVI) and the band sums."""
    i, folder, year, offset, kind, part = task
    try:
        out = COMPUTE[kind](folder, year)
    except Exception as e:
        return i, np.nan, None, f'{type(e).__name__}: {e}'[:200]
    if out is None:
        return i, np.nan, None, 'no-data pixels'
    x, value = out
    fd = os.open(part, os.O_WRONLY)
    try:
        os.pwrite(fd, x.astype(np.float16).tobytes(), offset)
    finally:
        os.close(fd)
    x64 = x.astype(np.float64)
    return i, value, np.stack([x64.sum(axis=(1, 2)), (x64 ** 2).sum(axis=(1, 2))]), ''


def build_cache(cache_dir, table, channels, kind, size=PATCH):
    """features.npy, band_stats.csv and index.csv of one cache folder for the clusters in table (format of 04-12;
    size: the pixels or cells per side)."""
    t0 = time.time()
    features = os.path.join(cache_dir, 'features.npy')
    shape = (len(table), len(channels), size, size)
    arr = np.lib.format.open_memmap(features + '.part', mode='w+', dtype=np.float16, shape=shape)
    header, per_cluster = arr.offset, int(np.prod(shape[1:])) * 2
    del arr
    tasks = [(i, r.folder, int(r.year), header + i * per_cluster, kind, features + '.part')
             for i, r in enumerate(table.itertuples())]
    value = np.full(len(table), np.nan)
    sums = np.zeros((2, len(channels)))
    error = [''] * len(table)
    with multiprocessing.get_context('fork').Pool(N_WORKERS) as pool:
        for i, t, s, err in tqdm(pool.imap_unordered(_cache_one, tasks, chunksize=4), total=len(tasks), desc=kind):
            value[i], error[i] = t, err
            if s is not None:
                sums += s
    ok = np.isfinite(value)
    n_pix = max(ok.sum(), 1) * size * size
    mean = sums[0] / n_pix
    std = np.sqrt(np.maximum(sums[1] / n_pix - mean ** 2, 1e-12))
    pd.DataFrame({'band': channels, 'mean': mean, 'std': std}).to_csv(os.path.join(cache_dir, 'band_stats.csv'),
                                                                      index=False)
    os.replace(features + '.part', features)
    index = table[['cluster_id', 'country', 'year', 'lat', 'lon']].copy()
    index[SUMMARY[kind]], index['ok'], index['error'] = value, ok, error
    index.to_csv(os.path.join(cache_dir, 'index.csv'), index=False)    # the cache counts as complete once this exists
    print(f'{kind} cache: {ok.sum()} of {len(ok)} clusters in {time.time() - t0:.0f} s -> {features}')


def cache_is_complete(cache_dir, table):
    index, features = os.path.join(cache_dir, 'index.csv'), os.path.join(cache_dir, 'features.npy')
    if not (os.path.exists(index) and os.path.exists(features)):
        return False
    return pd.read_csv(index)['cluster_id'].tolist() == table['cluster_id'].tolist()


def ensure_cache(cache_dir, table, channels, kind, size=PATCH):
    """Build a cache folder unless it is complete. Waits when another job is building it (same lock in 04 to 15 of both studies)."""
    os.makedirs(cache_dir, exist_ok=True)
    lock = os.path.join(cache_dir, 'building.lock')
    if cache_is_complete(cache_dir, table):
        print(f'{kind} cache found:', cache_dir)
        return
    try:
        os.mkdir(lock)
    except FileExistsError:
        print(f'another job is building {cache_dir} (if none is running, delete {lock}); waiting ...')
        t_wait = time.time()
        while os.path.exists(lock) and time.time() - t_wait < 4 * 3600:
            time.sleep(30)
        assert cache_is_complete(cache_dir, table), 'the cache is still not complete: start the notebook again'
        return
    try:
        if not cache_is_complete(cache_dir, table):     # another job may have finished it just now
            build_cache(cache_dir, table, channels, kind, size)
    finally:
        os.rmdir(lock)


def error_text(index):
    """The error column of a cache index as text ('' where there is none)."""
    e = index['error']
    return np.where(e.isna(), '', e.astype(str))


# ---- reading the 7 image bands, the 23 composites and the fixed normalisation (Settings) -------------------------
FROM_BANDS = np.array([BAND[b] for b in LANDSAT])                   # their positions in IMAGE ...
FROM_BANDS_CH = np.array([CHANNELS.index(b) for b in LANDSAT])      # ... and in the band cache


def read_input(pos):
    """Cached float16 image bands (IMAGE order: the 7 Landsat bands) of the clusters at the rows pos of `clusters`,
    shape (len(pos), 7, 224, 224)."""
    pos = np.asarray(pos)
    out = np.empty((len(pos), len(IMAGE), PATCH, PATCH), np.float16)
    part, prow = clusters['part'].to_numpy()[pos], clusters['part_row'].to_numpy()[pos]
    for s in np.unique(part):
        k = np.flatnonzero(part == s)
        k = k[np.argsort(prow[k])]                      # read in file order
        out[k[:, None], FROM_BANDS[None, :]] = FEAT[s][prow[k][:, None], FROM_BANDS_CH[None, :]]
    return out


def read_ndvi(pos):
    """Cached float16 NDVI composites of the clusters at the rows pos of `clusters`, shape (len(pos), 23, CELLS,
    CELLS)."""
    pos = np.asarray(pos)
    out = np.empty((len(pos), N_COMPOSITES, CELLS, CELLS), np.float16)
    part, prow = clusters['part'].to_numpy()[pos], clusters['part_row'].to_numpy()[pos]
    for s in np.unique(part):
        k = np.flatnonzero(part == s)
        k = k[np.argsort(prow[k])]                      # read in file order
        out[k] = NDF[s][prow[k]]
    return out


TEMP_FROM_CACHE = 300.0 / TEMP1_DIVISOR    # the band cache keeps TEMP1 in kelvin / 300


def normalise(x):
    """Training values of the image bands, in place (float32 tensor (n, 7, 224, 224), IMAGE order): Landsat with the
    GEE scale factors as cached (TEMP1 in kelvin / TEMP1_DIVISOR)."""
    if TEMP_FROM_CACHE != 1.0:
        x[:, BAND['TEMP1']] *= TEMP_FROM_CACHE
    return x


for s in NL_SENSORS:
    ensure_cache(BAND_DIRS[s], PARTS[s], CHANNELS, 'bands')
    ensure_cache(NDVI_DIRS[s], PARTS[s], COMPOSITES, 'composites', CELLS)
BAND_INDEX = {s: pd.read_csv(os.path.join(BAND_DIRS[s], 'index.csv')) for s in NL_SENSORS}
NDVI_INDEX = {s: pd.read_csv(os.path.join(NDVI_DIRS[s], 'index.csv')) for s in NL_SENSORS}
FEAT = {s: np.load(os.path.join(BAND_DIRS[s], 'features.npy'), mmap_mode='r') for s in NL_SENSORS}
NDF = {s: np.load(os.path.join(NDVI_DIRS[s], 'features.npy'), mmap_mode='r') for s in NL_SENSORS}
for s in NL_SENSORS:
    assert FEAT[s].shape == (len(PARTS[s]), len(CHANNELS), PATCH, PATCH), (s, FEAT[s].shape)
    assert NDF[s].shape == (len(PARTS[s]), N_COMPOSITES, CELLS, CELLS), (s, NDF[s].shape)
    print(f'{s} period: band cache {FEAT[s].shape}, {FEAT[s].nbytes / 1e9:.1f} GB; NDVI cache {NDF[s].shape}, '
          f'{NDF[s].nbytes / 1e9:.2f} GB')
complete = np.concatenate([BAND_INDEX[s]['ok'].to_numpy(bool) & NDVI_INDEX[s]['ok'].to_numpy(bool)
                           for s in NL_SENSORS])                  # (the band cache checks NL.tif too)
errors = pd.Series(np.concatenate([np.where(error_text(BAND_INDEX[s]) != '', error_text(BAND_INDEX[s]),
                                            error_text(NDVI_INDEX[s])) for s in NL_SENSORS]))
print(f'clusters with complete data: {complete.sum()} of {len(complete)}; left out:',
      errors[~complete].str.split(':').str[0].value_counts().to_dict())
assert complete.any(), 'no cluster has complete data'

# training values of every image band and composite for a sample of clusters of every period
rng = np.random.default_rng(0)
sample = np.sort(np.concatenate([rng.choice(c, size=min(16, len(c)), replace=False) for c in
                                 (np.flatnonzero(complete & (clusters['part'] == s).to_numpy()) for s in NL_SENSORS)
                                 if len(c)]))
x = normalise(torch.from_numpy(read_input(sample).astype(np.float32)))
nd = torch.from_numpy(read_ndvi(sample).astype(np.float32))
print(f'\ntraining values of the image bands and the composites (cells of {NDVI_POOL} x {NDVI_POOL} pixels), '
      f'{len(sample)} clusters:', clusters['part'].to_numpy()[sample].tolist().count('DMSP'), 'DMSP and',
      clusters['part'].to_numpy()[sample].tolist().count('VIIRS'), 'VIIRS period')
pd.DataFrame({'min': np.r_[x.amin(dim=(0, 2, 3)).numpy(), nd.amin(dim=(0, 2, 3)).numpy()],
              'mean': np.r_[x.mean(dim=(0, 2, 3)).numpy(), nd.mean(dim=(0, 2, 3)).numpy()],
              'max': np.r_[x.amax(dim=(0, 2, 3)).numpy(), nd.amax(dim=(0, 2, 3)).numpy()]},
             index=IMAGE + COMPOSITES).T.round(3)
````

# 4. Models

Every network has the same interface: `features(x, ndvi, loc)` gives the values of the layer before the output for a batch (`x`: the 7 normalised image bands, `ndvi`: the 23 composites of every cell, `loc`: the SatCLIP values), and `head`, a linear layer that runs in float32, turns them into the NL target. Inside the network, `stack()` runs the temporal transformer on `ndvi`, gives every pixel the 6 temporal bands of its cell and puts them behind the 7 image bands: the 13 bands of the table at the top, which then go through the networks of 11 unchanged. So all models read the same batches, and the temporal transformer learns with the rest of the network.

**The temporal transformer** (`TemporalEncoder`), for every cell, with the settings of the Settings cell:

1. **Tokens.** The 23 NDVI values, as (NDVI − 0.5) / 0.25, each through the same small MLP (1 → 64 → 64, GELU): one token of width 64 per composite. The token carries the NDVI value only; where it is in the year comes in step 2.
2. **Blocks.** `TEMPORAL_LAYERS` = 2 pre-norm transformer blocks: self-attention among the 23 tokens of the cell (4 heads of 16 channels), then an MLP (64 → 128 → 64), each with a residual connection. With `TEMPORAL_POSITION = 'cyclic'`, queries and keys are turned by the rotary position embedding (RoPE) before the attention: the channel pair j of the token of composite t is rotated by the angle 2π · j · t / 23, j = 0 … 7, the harmonics of the year. The attention weight between composites t and u then depends on their tokens and on (t − u) modulo 23 only, so the blocks are equivariant to cyclic shifts of the series. `'none'` leaves the tokens without any position (the blocks are then equivariant to every reordering), and `'absolute'` adds a learned vector for each of the 23 dates to the tokens before the first block.
3. **Pooling.** One learned query per head attends to the 23 tokens (attention pooling, as the master query of L-TAE). The query has no position, so the pooled vector does not depend on the order of the tokens; it starts at zero, which is the plain mean. With `cyclic` and `none`, the output of a cell is therefore invariant to shifting its series by whole composites; with `none` also to every reordering.
4. **Bands.** Layer norm, then a linear map to the `TEMPORAL_BANDS` = 6 temporal bands. They start small (a standard deviation of about 0.1 to 0.2, the scale of the Landsat reflectances) and are not otherwise scaled; every pixel of the cell gets them (nearest-neighbour upsampling, the way 01 put the MODIS pixels on the 30 m grid).

The linear layers start from a truncated normal (sd 0.02, as in ViT), the first layer of the token MLP from PyTorch's default, so that the 64 channels respond differently to the NDVI value. The cell after the next one checks the invariances on random series and prints the size of the transformer.

The networks are those of 11; only the input differs:

* **ResNet-18/34/50** (`ResNetRegressor`): torchvision's ResNet with a first convolution for the 13 bands, as the `finetune` mode of 05 and 07; global average pooling gives 512 (ResNet-50: 2048) features. `satclip` adds the 256 standardised SatCLIP values.
* **Swin v2-B** (`SwinRegressor`): torchgeo's `swin_v2_b` with the SatlasPretrain Landsat weights (single image). The patch embedding becomes an `ExtraProjection`: the pretrained 4 × 4 convolution on the 11 Level-1 bands made by `landsat_l1()`, plus a new 4 × 4 convolution of the `EXTRA` bands (TEMP1, NL and the temporal bands) that starts at zero, so the model starts as the pretrained one. torchvision's Swin v2-B has stochastic depth up to 0.5; it is scaled to `DROP_PATH`. Features: the 1024 values after the final norm and average pooling.
* **U-Net** (`UNetRegressor`): torchgeo's `unet` (segmentation_models_pytorch) with the FTW weights. Its first convolution gets the 13 bands (RED, GREEN, BLUE, NIR: the weights of B4, B3, B2, B8 of both dates added, which is the same image at both dates; the other bands, the temporal ones included, start at zero). The segmentation head is replaced by the 16 decoder channels averaged over the image and standardised (BatchNorm without learned scale) and `head`. In evaluation mode this is the same as a linear map of the 16 decoder channels to an NL value per pixel, averaged over the image; `light_map()` gives that map.
* **ViT-S/16** (`ViTRegressor`): as `LandsatViT` of 08, built with torchgeo's `vit_small_patch16_224`: the class token after the final norm, 384 features.
* **SatCLIP** (`satclip_embeddings()`, `LocationMLP`): torchgeo's `satclip` location encoder turns the longitude and latitude of every cluster into 256 values once (section 5); they are standardised with the mean and standard deviation of the training clusters of each model. `satclip_only` feeds them to two hidden layers of 256 (ReLU), without image or NDVI.
* **MOSAIKS** (`run_mosaiks()`, the last cell of this section): no temporal transformer, because nothing in MOSAIKS learns by gradient. Its 30 bands are the 7 image bands and the 23 composites (every pixel with the values of its cell), each standardised with the mean and standard deviation of the training clusters' pixels. torchgeo's `RCF` (empirical, as its `MOSAIKS` class: kernel 4, bias −1) takes 2048 random 4 × 4 × 30 patches of training images, whitened (ZCA), as filters, and every image gives 4096 features (the mean of the positive and of the negative part of each filter response). A ridge regression (features standardised, alpha from `ALPHAS` with the lowest MSE on the validation fold, then fitted on the training and validation folds) predicts the target. Filters and scaling come from the training folds only, so the test fold is never seen.
* **OlmoEarth v1** (`OlmoEarthRegressor`): torchgeo's `olmoearth_v1`; only its encoder is kept. The Landsat bands go in as the `landsat` modality, one time step (timestamp July), the bands of `landsat_l1()` normalised as in OlmoEarth's pretraining, with `input_res` = 30 m. `TokensWithExtra` adds a new projection of the `EXTRA` bands (8 × 8 convolution, starting at zero) to the Landsat tokens; with `NDVI_POOL = 8` each token covers exactly one cell of the temporal transformer. Features: the mean of the encoder's output tokens (768 for Base).
* **VGG-16-BN** (`VGGRegressor`, baseline): torchvision's `vgg16_bn` with `VGG16_BN_Weights.IMAGENET1K_V1` (or random weights) after the temporal transformer (`stack`); its first 3 × 3 convolution takes the input bands like the ResNets' first convolution; global average pooling after the last block gives 512 features; the ImageNet classifier is not used.
* **MLP** (`BandStatsMLP`, baseline): no convolution after the temporal transformer (`stack`); for every input band, the mean, standard deviation and `MLP_QUANTILES` percentiles of its pixels, standardised by BatchNorm and fed to the `MLP_HIDDEN` hidden layers (ReLU): 256 features.

A run trains the model of one test fold. After every epoch it writes the history (with the learning rates), the best weights (`best.pt`, lowest validation loss) and its full state (`last.pt`), both with the settings of the temporal transformer. Then it loads `best.pt`, computes the prediction (and the features) of every cluster in float32 and writes `predictions.csv`, `features.csv`, `test_predictions.csv` and `result.json`, and deletes `last.pt`, which is only needed to continue a stopped run. A run with `result.json` (and `features.csv`, with `SAVE_FEATURES`) is skipped. A run with only a `last.pt` continues from that epoch. A finished run whose `features.csv` was deleted writes it again from `best.pt`. A run folder holds runs with one start and one temporal transformer: `check_start()` stops the notebook when the settings of a run it finds differ from those of Settings.

````python
# ---- pretrained weights: torchvision and torchgeo read them from TORCH_HOME/hub/checkpoints ------------------------
CKPT_DIR = os.path.join(TORCH_HOME, 'hub', 'checkpoints')
OLMOEARTH_HINT = ('OlmoEarth needs olmoearth_pretrain_minimal. Install it once:\n  '
                  'pip install --no-deps --target ' + PY_PACKAGES +
                  ' olmoearth-pretrain-minimal==0.1.5 universal-pathlib pathlib-abc')


def tg_weights(enum_name, member):
    """A weights entry of torchgeo (0.10 or later)."""
    import torchgeo.models
    return getattr(getattr(torchgeo.models, enum_name), member)


def weights_of(key):
    """The pretrained weights (objects with .url) a model reads in this run."""
    tv = torchvision.models
    resnet = {18: tv.ResNet18_Weights.IMAGENET1K_V1, 34: tv.ResNet34_Weights.IMAGENET1K_V1,
              50: tv.ResNet50_Weights.IMAGENET1K_V1}
    need = []
    if PRETRAINED:
        if key in ('resnet18', 'resnet34', 'resnet50'):
            need.append(resnet[int(key[6:])])
        elif key == 'satclip':
            need.append(resnet[18])
        elif key == 'swin':
            need.append(tg_weights('Swin_V2_B_Weights', 'LANDSAT_SI_SATLAS'))
        elif key == 'unet':
            need.append(tg_weights('Unet_Weights', 'SENTINEL2_2CLASS_FTW'))
        elif key == 'vit':
            need.append(tg_weights('ViTSmall16_Weights', 'LANDSAT_ETM_SR_MOCO'))
        elif key == 'olmoearth':
            need.append(tg_weights('OlmoEarthV1_Weights', OLMOEARTH_SIZE))
        elif key == 'vgg16bn':
            need.append(tv.VGG16_BN_Weights.IMAGENET1K_V1)
    if 'location' in MODELS[key][3]:          # the SatCLIP values are an input, also with TRAIN_FROM_SCRATCH
        need.append(tg_weights('SatCLIP_Weights', 'SATCLIP_VIT16_L40'))
    return need


def download_hint(key):
    try:
        urls = [w.url for w in weights_of(key)]
    except Exception as e:
        return f'(the weights of {key} could not be listed: {type(e).__name__}: {e})'
    return ('If this machine has no internet access, download the weights first, then run again:\n  mkdir -p '
            + CKPT_DIR + ''.join(f'\n  wget -O {os.path.join(CKPT_DIR, os.path.basename(u))} {u}' for u in urls))


# ---- Landsat for the models pretrained on Landsat 8/9 Level-1 data (Swin, OlmoEarth) ---------------------------
L1_BANDS = ['B1', 'B2', 'B3', 'B4', 'B5', 'B6', 'B7', 'B8', 'B9', 'B10', 'B11']
TIRS = ((774.8853, 1321.0789), (480.8883, 1201.1442))   # K1, K2 of Landsat 8 TIRS B10 and B11 (MTL files)
CIRRUS_DN = 5000.0                                     # B9 (cirrus) of a clear sky: TOA reflectance about 0
SATLAS_LANDSAT = (4000.0, 16320.0)                     # SatlasPretrain: clip((DN - 4000) / 16320, 0, 1)


def landsat_l1(x):
    """Approximate Landsat 8/9 Level-1 digital numbers of B1-B11, (n, 11, H, W), from the normalised input bands x
    (n, 13, H, W): surface reflectance taken as TOA reflectance, DN = (reflectance + 0.1) / 2e-5; B1 (coastal) = BLUE,
    B8 (panchromatic) = mean of GREEN and RED, B9 (cirrus) = CIRRUS_DN; B10 and B11 = the radiance of TEMP1 (Planck,
    TIRS constants) as DN = (L - 0.1) / 3.342e-4."""
    blue, green, red, nir, swir1, swir2 = ((x[:, :6] + 0.1) / 2e-5).unbind(1)
    kelvin = (x[:, BAND['TEMP1']] * TEMP1_DIVISOR).clamp(min=150.0)
    b10, b11 = [(k1 / torch.expm1(k2 / kelvin) - 0.1) / 3.342e-4 for k1, k2 in TIRS]
    return torch.stack([blue, blue, green, red, nir, swir1, swir2, (green + red) / 2, torch.full_like(blue, CIRRUS_DN),
                        b10, b11], dim=1)


def swin_input(x):
    """The 11 bands of the SatlasPretrain Landsat weights, normalised as in SatlasPretrain."""
    return ((landsat_l1(x) - SATLAS_LANDSAT[0]) / SATLAS_LANDSAT[1]).clamp(0, 1)


def vit_input(x):
    """The six SR bands of the SSL4EO-L weights (ETM+ SR_B1-B5, B7), as in 08: clipped to 0-SR_MAX, / SR_MAX."""
    return (x[:, :6] / SR_MAX).clamp(0, 1)


# ---- the temporal transformer: the 23 composites of every cell -> TEMPORAL_BANDS learned bands -----------------------
def cyclic_rope_tables(n_steps, head_dim, period):
    """cos and sin, (n_steps, head_dim // 2), of the rotary position embedding with the harmonics 0, 1, ...,
    head_dim / 2 - 1 of one period: channel pair j of step t is turned by the angle 2 pi j t / period. With
    period = n_steps the angles of t and t + period are the same, so the attention only sees (t - u) mod period."""
    j = torch.arange(head_dim // 2, dtype=torch.float64)
    t = torch.arange(n_steps, dtype=torch.float64)
    angle = 2 * math.pi * torch.outer(t, j) / period
    return torch.cos(angle), torch.sin(angle)             # float64; rotate() casts them to the dtype of x


def rotate(x, cos, sin):
    """Rotary embedding of x (..., n_steps, head_dim): every channel pair (2j, 2j + 1) turned by its angle."""
    cos, sin = cos.to(x.dtype), sin.to(x.dtype)
    x1, x2 = x[..., 0::2], x[..., 1::2]
    return torch.stack([x1 * cos - x2 * sin, x1 * sin + x2 * cos], dim=-1).flatten(-2)


class TemporalBlock(nn.Module):
    """Pre-norm transformer block over the tokens of every cell: self-attention (with RoPE when rope is given), then
    an MLP, each with a residual connection."""

    def __init__(self, dim, heads, mlp_ratio, dropout):
        super().__init__()
        self.heads, self.dropout = heads, dropout
        self.norm1 = nn.LayerNorm(dim)
        self.qkv = nn.Linear(dim, 3 * dim)
        self.proj = nn.Linear(dim, dim)
        self.norm2 = nn.LayerNorm(dim)
        self.mlp = nn.Sequential(nn.Linear(dim, mlp_ratio * dim), nn.GELU(), nn.Dropout(dropout),
                                 nn.Linear(mlp_ratio * dim, dim))

    def forward(self, z, rope=None):
        n, t, d = z.shape
        q, k, v = self.qkv(self.norm1(z)).view(n, t, 3, self.heads, d // self.heads).permute(2, 0, 3, 1, 4).unbind(0)
        if rope is not None:                            # (n, heads, t, head_dim) each
            q, k = rotate(q, *rope), rotate(k, *rope)
        a = F.scaled_dot_product_attention(q, k, v, dropout_p=self.dropout if self.training else 0.0)
        z = z + self.proj(a.transpose(1, 2).reshape(n, t, d))
        return z + self.mlp(self.norm2(z))


class AttentionPool(nn.Module):
    """One learned query per head attends to the tokens of a cell: a weighted mean of the tokens. The query has no
    position, so the result does not depend on the order of the tokens. It starts at zero: the plain mean."""

    def __init__(self, dim, heads):
        super().__init__()
        self.heads = heads
        self.norm = nn.LayerNorm(dim)
        self.kv = nn.Linear(dim, 2 * dim)
        self.query = nn.Parameter(torch.zeros(heads, dim // heads))

    def forward(self, z):
        n, t, d = z.shape
        k, v = self.kv(self.norm(z)).view(n, t, 2, self.heads, d // self.heads).unbind(2)    # (n, t, heads, head_dim)
        w = torch.einsum('nthc,hc->nht', k, self.query.to(k.dtype)) / math.sqrt(d // self.heads)
        return torch.einsum('nht,nthc->nhc', w.softmax(dim=-1).to(v.dtype), v).reshape(n, d)


class TemporalEncoder(nn.Module):
    """The temporal transformer: NDVI composites of every cell, (n, 23, h, w), -> TEMPORAL_BANDS bands (n, bands, h, w).
    One token per composite (its NDVI value through a small MLP), TEMPORAL_LAYERS blocks of self-attention among the 23
    tokens of a cell, attention pooling, layer norm and a linear map. position: 'cyclic' (RoPE with the harmonics of
    the year), 'none' or 'absolute' (a learned vector per date), see Settings."""

    def __init__(self, position=None):
        super().__init__()
        self.position = position or TEMPORAL_POSITION
        assert self.position in POSITIONS, self.position
        d, h = TEMPORAL_DIM, TEMPORAL_HEADS
        self.embed = nn.Sequential(nn.Linear(1, d), nn.GELU(), nn.Linear(d, d))
        self.pos = nn.Parameter(torch.zeros(N_COMPOSITES, d)) if self.position == 'absolute' else None
        if self.position == 'cyclic':                   # 23 composites = one period
            cos, sin = cyclic_rope_tables(N_COMPOSITES, d // h, N_COMPOSITES)
            self.register_buffer('rope_cos', cos, persistent=False)
            self.register_buffer('rope_sin', sin, persistent=False)
        self.blocks = nn.ModuleList(TemporalBlock(d, h, TEMPORAL_MLP, TEMPORAL_DROPOUT)
                                    for _ in range(TEMPORAL_LAYERS))
        self.pool = AttentionPool(d, h)
        self.norm = nn.LayerNorm(d)
        self.out = nn.Linear(d, TEMPORAL_BANDS)
        for m in self.modules():                        # as in ViT: linear layers N(0, 0.02), biases 0
            if isinstance(m, nn.Linear):
                nn.init.trunc_normal_(m.weight, std=0.02)
                nn.init.zeros_(m.bias)
        self.embed[0].reset_parameters()                # PyTorch's default: the 64 channels see the value differently
        if self.pos is not None:
            nn.init.trunc_normal_(self.pos, std=0.02)

    def encode(self, v):
        """NDVI series v (cells, 23) -> bands (cells, TEMPORAL_BANDS)."""
        z = self.embed(((v - NDVI_CENTER) / NDVI_SCALE).unsqueeze(-1))
        if self.pos is not None:
            z = z + self.pos
        rope = (self.rope_cos, self.rope_sin) if self.position == 'cyclic' else None
        for block in self.blocks:
            z = block(z, rope)
        return self.out(self.norm(self.pool(z)))

    def forward(self, ndvi):
        n, t, h, w = ndvi.shape
        assert t == N_COMPOSITES, ndvi.shape
        v = ndvi.permute(0, 2, 3, 1).reshape(n * h * w, t)
        out = torch.cat([self.encode(v[s:s + TEMPORAL_CHUNK]) for s in range(0, len(v), TEMPORAL_CHUNK)])
        return out.view(n, h, w, -1).permute(0, 3, 1, 2)


# ---- the models ---------------------------------------------------------------------------------------------------
class Regressor(nn.Module):
    """Common part of the networks: features(x, ndvi, loc) -> (n, n_features), and `head`, the linear output in
    float32. The networks that read the NDVI series have a `temporal` transformer (make_temporal), and stack() makes
    their 13 input bands."""

    def __init__(self):
        super().__init__()
        self.temporal = None                            # make_temporal() sets it in the networks that read the NDVI

    def make_head(self, n_features):
        self.n_features = n_features
        self.head = nn.Linear(n_features, 1)
        nn.init.xavier_uniform_(self.head.weight)              # Keras defaults, as in 05-08; the bias is set to the
        nn.init.zeros_(self.head.bias)                         # mean target of the training clusters in run_fold

    def make_temporal(self):
        """The temporal transformer of this network (random weights, also when the rest is pretrained)."""
        self.temporal = TemporalEncoder()

    def stack(self, x, ndvi):
        """The 13 input bands (INPUT order) of a batch: the 7 image bands x (n, 7, 224, 224) and the TEMPORAL_BANDS
        bands that the temporal transformer makes of the composites ndvi (n, 23, CELLS, CELLS), every pixel with the
        bands of its cell."""
        t = self.temporal(ndvi)
        if t.shape[-2:] != x.shape[-2:]:
            t = F.interpolate(t, size=x.shape[-2:], mode='nearest')
        return torch.cat([x, t.to(x.dtype)], dim=1).contiguous(memory_format=torch.channels_last)

    def forward(self, x, ndvi=None, loc=None):
        f = self.features(x, ndvi, loc)
        with torch.autocast(device_type=f.device.type, enabled=False):    # output layer in float32
            return self.head(f.float()).squeeze(1)

    @staticmethod
    def is_new(name):
        """Parameters that are not pretrained (adamw: LR_NEW): the output layer, the projection of the EXTRA bands and
        the temporal transformer."""
        return name.startswith(('head.', 'temporal.')) or '.extra.' in name

    def set_location_scale(self, loc):
        """Standardise the SatCLIP values with the mean and standard deviation of the training clusters (loc)."""
        self.loc_mean.copy_(loc.mean(dim=0))
        self.loc_std.copy_(loc.std(dim=0).clamp(min=1e-6))


def first_conv_weights(bands, rgb_weights):
    """First-layer weights for any bands from the ImageNet ones (Yeh et al. 2020): RED, GREEN and BLUE keep their
    own, every other band gets their mean, and all are scaled by 3 / number of bands (05, 07)."""
    rgb_index = {'RED': 0, 'GREEN': 1, 'BLUE': 2}            # channel order of the ImageNet weights
    mean = rgb_weights.mean(dim=1)
    w = torch.stack([rgb_weights[:, rgb_index[b]] if b in rgb_index else mean for b in bands], dim=1)
    return w * 3.0 / len(bands)


LOC_DIM = 256                                                  # SatCLIP values per location


class ResNetRegressor(Regressor):
    """torchvision ResNet-18/34/50 whose first convolution takes the 13 bands (the finetune mode of 05 and 07): the 7
    image bands and the temporal bands. location=True (satclip): the 256 standardised SatCLIP values join the pooled
    features before the output."""

    def __init__(self, depth, location=False):
        super().__init__()
        net = getattr(torchvision.models, f'resnet{depth}')(weights='IMAGENET1K_V1' if PRETRAINED else None)
        rgb_weights = net.conv1.weight.detach().clone()        # (64, 3, 7, 7)
        net.conv1 = nn.Conv2d(len(INPUT), 64, kernel_size=7, stride=2, padding=3, bias=False)
        with torch.no_grad():
            if PRETRAINED:
                net.conv1.weight.copy_(first_conv_weights(INPUT, rgb_weights))
            else:                                              # torchvision's initialisation of its convolutions
                nn.init.kaiming_normal_(net.conv1.weight, mode='fan_out', nonlinearity='relu')
        n_features = net.fc.in_features
        net.fc = nn.Identity()                                 # the network now ends with the pooled features
        self.backbone = net
        self.make_temporal()
        self.location = location
        if location:
            self.register_buffer('loc_mean', torch.zeros(LOC_DIM))
            self.register_buffer('loc_std', torch.ones(LOC_DIM))
            n_features += LOC_DIM
        self.make_head(n_features)

    def features(self, x, ndvi=None, loc=None):
        f = self.backbone(self.stack(x, ndvi))
        if self.location:
            f = torch.cat([f, ((loc - self.loc_mean) / self.loc_std).to(f.dtype)], dim=1)
        return f


class LocationMLP(Regressor):
    """satclip_only: the standardised SatCLIP values of the cluster's coordinates, two hidden layers (ReLU)."""

    def __init__(self, hidden=256):
        super().__init__()
        self.register_buffer('loc_mean', torch.zeros(LOC_DIM))
        self.register_buffer('loc_std', torch.ones(LOC_DIM))
        self.mlp = nn.Sequential(nn.Linear(LOC_DIM, hidden), nn.ReLU(), nn.Linear(hidden, hidden), nn.ReLU())
        self.make_head(hidden)

    def features(self, x, ndvi, loc):
        return self.mlp((loc - self.loc_mean) / self.loc_std)


class VGGRegressor(Regressor):
    """vgg16bn (baseline): torchvision's VGG-16 with batch normalisation. Its first convolution takes all input bands,
    the 7 image bands and the temporal bands of the temporal transformer, like the ResNets: RED, GREEN and BLUE keep
    their ImageNet weights, every other band gets their mean, all scaled by 3 / number of bands. The convolution layers
    end in global average pooling: 512 features (the 4096-wide ImageNet classifier is not used, as the ResNets' fc
    layer)."""

    def __init__(self):
        super().__init__()
        net = torchvision.models.vgg16_bn(weights='IMAGENET1K_V1' if PRETRAINED else None)
        conv = net.features[0]                                 # Conv2d(3, 64, 3, padding=1)
        first = nn.Conv2d(len(INPUT), conv.out_channels, kernel_size=3, padding=1)
        with torch.no_grad():
            if PRETRAINED:
                first.weight.copy_(first_conv_weights(INPUT, conv.weight.detach()))
                first.bias.copy_(conv.bias.detach())
            else:                                              # torchvision's initialisation of the VGG convolutions
                nn.init.kaiming_normal_(first.weight, mode='fan_out', nonlinearity='relu')
                nn.init.zeros_(first.bias)
        net.features[0] = first
        self.backbone = nn.Sequential(net.features, nn.AdaptiveAvgPool2d(1), nn.Flatten())
        self.make_temporal()
        self.make_head(512)

    def features(self, x, ndvi=None, loc=None):
        return self.backbone(self.stack(x, ndvi))


class BandStatsMLP(Regressor):
    """mlp (baseline without convolutions): every input band of the patch is summarised by its mean, its standard
    deviation and its MLP_QUANTILES percentiles over the 224 x 224 pixels. These values are standardised (BatchNorm,
    learned from the training batches) and go through the MLP_HIDDEN hidden layers (ReLU). The MLP sees what the
    bands of a patch contain, but not where: its gap to the CNNs measures what the spatial patterns add."""

    def __init__(self):
        super().__init__()
        n_in = len(INPUT) * (2 + len(MLP_QUANTILES))
        self.norm = nn.BatchNorm1d(n_in)
        layers, width = [], n_in
        for h in MLP_HIDDEN:
            layers += [nn.Linear(width, h), nn.ReLU()]
            width = h
        self.mlp = nn.Sequential(*layers)
        self.make_temporal()
        self.make_head(width)

    @staticmethod
    def band_statistics(x):
        """(n, C, H, W) -> (n, C * (2 + len(MLP_QUANTILES))), float32: the mean and standard deviation of every band,
        then its percentiles MLP_QUANTILES (the sorted pixel values at the nearest rank)."""
        v = x.float().flatten(2)                               # (n, C, H * W)
        last = v.shape[2] - 1
        ranks = [min(max(int(round(q * last)), 0), last) for q in MLP_QUANTILES]
        q = v.sort(dim=2).values[:, :, ranks]                  # (n, C, len(MLP_QUANTILES))
        return torch.cat([v.mean(dim=2), v.std(dim=2), q.flatten(1)], dim=1)

    def features(self, x, ndvi=None, loc=None):
        x = self.stack(x, ndvi)
        with torch.autocast(device_type=x.device.type, enabled=False):    # statistics and BatchNorm in float32
            z = self.norm(self.band_statistics(x))
        return self.mlp(z)


class ExtraProjection(nn.Module):
    """A pretrained patch embedding (a convolution) fed by `prepare` (input bands -> the bands of the weights), plus a
    new convolution of the same shape for the EXTRA bands, without bias and starting at zero, so that the model
    starts as the pretrained one (TRAIN_FROM_SCRATCH: both with PyTorch's random initialisation)."""

    def __init__(self, pretrained, prepare):
        super().__init__()
        self.pretrained = pretrained
        self.prepare = prepare
        self.extra = nn.Conv2d(len(EXTRA), pretrained.out_channels, kernel_size=pretrained.kernel_size,
                               stride=pretrained.stride, padding=pretrained.padding, bias=False)
        if PRETRAINED:
            nn.init.zeros_(self.extra.weight)

    def forward(self, x):
        return self.pretrained(self.prepare(x)) + self.extra(x[:, EXTRA_INDEX])


class SwinRegressor(Regressor):
    """torchgeo's Swin v2-B with the SatlasPretrain Landsat weights (single image), 1024 features."""

    def __init__(self):
        super().__init__()
        from torchgeo.models import Swin_V2_B_Weights, swin_v2_b
        from torchvision.ops import StochasticDepth
        net = swin_v2_b(Swin_V2_B_Weights.LANDSAT_SI_SATLAS if PRETRAINED else None)
        if not PRETRAINED:                                     # torchgeo sets the 11-band embedding only with weights
            net.features[0][0] = nn.Conv2d(len(L1_BANDS), net.features[0][0].out_channels, kernel_size=4, stride=4)
        assert net.features[0][0].in_channels == len(L1_BANDS), net.features[0][0]
        net.features[0][0] = ExtraProjection(net.features[0][0], swin_input)
        depth = [m for m in net.modules() if isinstance(m, StochasticDepth)]
        top = max(m.p for m in depth)                          # torchvision: 0 ... 0.5 over the blocks
        for m in depth:
            m.p = m.p / top * DROP_PATH
        n_features = net.head.in_features
        net.head = nn.Identity()                               # features -> norm -> average pooling: 1024 values
        self.backbone = net
        self.make_temporal()
        self.make_head(n_features)

    def features(self, x, ndvi=None, loc=None):
        x = self.stack(x, ndvi)
        if SWIN_SIZE != x.shape[-1]:
            x = F.interpolate(x, size=(SWIN_SIZE, SWIN_SIZE), mode='bilinear', align_corners=False)
        return self.backbone(x)


class UNetRegressor(Regressor):
    """torchgeo's U-Net (segmentation_models_pytorch, EfficientNet-B3 encoder) with the FTW weights. The 16 decoder
    channels are averaged over the image and standardised (`pool_norm`: BatchNorm without learned scale, so that the
    output layer starts on a useful scale; the averages of BatchNorm + ReLU maps hardly differ between images): these
    are the 16 features. In evaluation mode this is the mean over the image of an NL value per pixel (light_map())."""

    FTW_BANDS = ['RED', 'GREEN', 'BLUE', 'NIR']                # B4, B3, B2, B8, at each of the two dates

    def __init__(self):
        super().__init__()
        from torchgeo.models import Unet_Weights, unet
        net = unet(Unet_Weights.SENTINEL2_2CLASS_FTW if PRETRAINED else None, encoder_name='efficientnet-b3',
                   in_channels=2 * len(self.FTW_BANDS), classes=2)
        stem = next(m for m in net.encoder.modules() if isinstance(m, nn.Conv2d))
        assert stem.in_channels == 2 * len(self.FTW_BANDS), stem
        w = stem.weight.detach()                               # (40, 8, 3, 3)
        new = torch.zeros(w.shape[0], len(INPUT), *w.shape[2:])
        if PRETRAINED:                                         # the same image at both dates
            for j, b in enumerate(self.FTW_BANDS):
                new[:, BAND[b]] = w[:, j] + w[:, j + len(self.FTW_BANDS)]
        else:
            nn.init.kaiming_normal_(new, mode='fan_out', nonlinearity='relu')
        stem.weight = nn.Parameter(new)
        stem.in_channels = len(INPUT)
        n_features = net.segmentation_head[0].in_channels       # 16
        net.segmentation_head = nn.Identity()                  # net(x) is then the decoder output
        self.net = net
        self.register_buffer('ftw_index', torch.tensor([BAND[b] for b in self.FTW_BANDS]), persistent=False)
        self.pool_norm = nn.BatchNorm1d(n_features, affine=False)
        self.make_temporal()
        self.make_head(n_features)

    def decoded(self, x):
        x = x.clone()
        x[:, self.ftw_index] = x[:, self.ftw_index] / FTW_SCALE    # as Sentinel-2 DN / 3000
        return self.net(x)                                     # (n, 16, 224, 224)

    def features(self, x, ndvi=None, loc=None):
        return self.pool_norm(self.decoded(self.stack(x, ndvi)).mean(dim=(2, 3)))

    def light_map(self, x, ndvi):
        """The NL target per pixel, (n, 224, 224), in evaluation mode; its mean over the image is the prediction."""
        bn = self.pool_norm
        scale = (bn.running_var + bn.eps).sqrt().view(1, -1, 1, 1)
        z = (self.decoded(self.stack(x, ndvi)).float() - bn.running_mean.view(1, -1, 1, 1)) / scale
        return torch.einsum('nchw,c->nhw', z, self.head.weight[0].float()) + self.head.bias.float()


class ViTRegressor(Regressor):
    """ViT-S/16 with the SSL4EO-L MoCo weights (LandsatViT of 08): the class token after the final norm, 384 values."""

    def __init__(self):
        super().__init__()
        from torchgeo.models import ViTSmall16_Weights, vit_small_patch16_224
        net = vit_small_patch16_224(ViTSmall16_Weights.LANDSAT_ETM_SR_MOCO if PRETRAINED else None,
                                    in_chans=len(SR_BANDS), num_classes=0, drop_path_rate=DROP_PATH)
        net.patch_embed.proj = ExtraProjection(net.patch_embed.proj, vit_input)
        self.backbone = net
        self.make_temporal()
        self.make_head(net.num_features)

    def features(self, x, ndvi=None, loc=None):
        return self.backbone(self.stack(x, ndvi))


class TokensWithExtra(nn.Module):
    """OlmoEarth's patch embeddings plus a new projection of the EXTRA bands, added to the Landsat tokens of every band
    set; it starts at zero (TRAIN_FROM_SCRATCH: random). OlmoEarthRegressor hands over the bands in `extra_input`."""

    def __init__(self, inner, dim, patch):
        super().__init__()
        self.inner = inner
        self.extra = nn.Conv2d(len(EXTRA), dim, kernel_size=patch, stride=patch, bias=False)
        if PRETRAINED:
            nn.init.zeros_(self.extra.weight)
        self.extra_input = None

    def forward(self, input_data, patch_size):
        out = self.inner.forward(input_data, patch_size)
        if self.extra_input is not None:
            e = self.extra(self.extra_input).permute(0, 2, 3, 1)         # (n, h, w, dim)
            out['landsat'] = out['landsat'] + e[:, :, :, None, None, :].to(out['landsat'].dtype)
        return out


class OlmoEarthRegressor(Regressor):
    """The encoder of OlmoEarth v1 on the Landsat bands (its `landsat` modality, one time step); features: the mean of
    its output tokens (768 values for Base)."""

    def __init__(self):
        super().__init__()
        from torchgeo.models import OlmoEarthV1_Weights, olmoearth_v1
        from olmoearth_pretrain_minimal import Normalizer
        from olmoearth_pretrain_minimal.olmoearth_pretrain_v1.utils.constants import Modality
        from olmoearth_pretrain_minimal.olmoearth_pretrain_v1.utils.datatypes import MaskedOlmoEarthSample, MaskValue
        from olmoearth_pretrain_minimal.olmoearth_pretrain_v1.nn.flexi_vit import PoolingType
        self._sample, self._pooling = MaskedOlmoEarthSample, PoolingType.MEAN
        self._online, self._missing = MaskValue.ONLINE_ENCODER.value, MaskValue.MISSING.value
        full = olmoearth_v1(getattr(OlmoEarthV1_Weights, OLMOEARTH_SIZE) if PRETRAINED else None,
                            model_size=OLMOEARTH_SIZE.lower(), drop_path=DROP_PATH)
        encoder = full.model.encoder                           # the decoder and target of pretraining are not needed
        del full
        order = Modality.LANDSAT.band_order                    # ['B8', 'B1', ..., 'B7', 'B9', 'B10', 'B11']
        stats = Normalizer(std_multiplier=2.0).norm_config['landsat']    # OlmoEarth: mean +- 2 sd -> 0 ... 1
        mean = torch.tensor([stats[b]['mean'] for b in order], dtype=torch.float32)
        std = torch.tensor([stats[b]['std'] for b in order], dtype=torch.float32)
        self.register_buffer('l1_order', torch.tensor([L1_BANDS.index(b) for b in order]), persistent=False)
        self.register_buffer('lo', (mean - 2 * std).view(1, -1, 1, 1), persistent=False)
        self.register_buffer('span', (4 * std).view(1, -1, 1, 1), persistent=False)
        groups = encoder.patch_embeddings.tokenization_config.get_bandset_indices('landsat')
        self.n_sets = len(groups)
        self.pan_sets = [k for k, g in enumerate(groups) if list(g) == [order.index('B8')]]   # band sets of B8 alone
        encoder.patch_embeddings = TokensWithExtra(encoder.patch_embeddings, encoder.embedding_size, OLMOEARTH_PATCH)
        self.encoder = encoder
        self.make_temporal()
        self.make_head(encoder.embedding_size)

    def sample(self, x):
        """The OlmoEarth input of a batch: landsat (n, H, W, 1, 11), its mask (n, H, W, 1, band sets), timestamps."""
        n, _, h, w = x.shape
        landsat = ((landsat_l1(x)[:, self.l1_order] - self.lo) / self.span).permute(0, 2, 3, 1).unsqueeze(3)
        mask = torch.full((n, h, w, 1, self.n_sets), self._online, dtype=torch.long, device=x.device)
        if OLMOEARTH_PAN == 'missing':
            mask[..., self.pan_sets] = self._missing
        timestamps = torch.tensor([15, 6, 2018], device=x.device).expand(n, 1, 3)   # day, month (0 = Jan), year
        return self._sample(timestamps=timestamps, landsat=landsat.contiguous(), landsat_mask=mask)

    def features(self, x, ndvi=None, loc=None):
        x = self.stack(x, ndvi)
        tokens = self.encoder.patch_embeddings
        tokens.extra_input = x[:, EXTRA_INDEX]
        try:
            out = self.encoder(self.sample(x), patch_size=OLMOEARTH_PATCH, input_res=30,
                               fast_pass=OLMOEARTH_PAN == 'copy')    # with masked tokens, the slower path
        finally:
            tokens.extra_input = None
        return out['tokens_and_masks'].pool_unmasked_tokens(self._pooling, spatial_pooling=False)


def build_model(key):
    """A new network for a model key, with its pretrained weights (TRAIN_FROM_SCRATCH: random)."""
    try:
        if key in ('resnet18', 'resnet34', 'resnet50'):
            return ResNetRegressor(int(key[6:]))
        if key == 'satclip':
            return ResNetRegressor(18, location=True)
        if key == 'satclip_only':
            return LocationMLP()
        if key == 'swin':
            return SwinRegressor()
        if key == 'unet':
            return UNetRegressor()
        if key == 'vit':
            return ViTRegressor()
        if key == 'olmoearth':
            return OlmoEarthRegressor()
        if key == 'vgg16bn':
            return VGGRegressor()
        if key == 'mlp':
            return BandStatsMLP()
    except ImportError as e:
        raise RuntimeError(f'{key}: a package is missing ({e}).' + ('\n' + OLMOEARTH_HINT if key == 'olmoearth' else
                                                                    '')) from e
    except Exception as e:
        raise RuntimeError(f'{key} could not be built ({type(e).__name__}: {e}).\n{download_hint(key)}') from e
    raise ValueError(f'{key} is not a network')


def satclip_embeddings(pos):
    """The 256 values of torchgeo's SatCLIP location encoder (SATCLIP_VIT16_L40) for the longitude and latitude of the
    clusters at the rows pos of `clusters` (float32, on DEVICE)."""
    try:
        from torchgeo.models import SatCLIP_Weights, satclip
        encoder = satclip(SatCLIP_Weights.SATCLIP_VIT16_L40)
    except Exception as e:
        raise RuntimeError(f'the SatCLIP encoder could not be built ({type(e).__name__}: {e}).\n'
                           f'{download_hint("satclip_only")}') from e
    encoder = encoder.eval().to(DEVICE)
    lonlat = torch.tensor(clusters[['lon', 'lat']].to_numpy(np.float64)[np.asarray(pos)], dtype=torch.float32)
    with torch.no_grad():
        emb = torch.cat([encoder(lonlat[s:s + 4096].to(DEVICE)).float() for s in range(0, len(lonlat), 4096)])
    assert emb.shape == (len(lonlat), LOC_DIM), emb.shape
    return emb


# the weights files of the models to run
missing = {}
for key in todo:
    try:
        for w in weights_of(key):
            f = os.path.join(CKPT_DIR, os.path.basename(w.url))
            if not os.path.exists(f):
                missing[f] = w.url
    except Exception as e:
        print(f'{key}: the weights could not be listed ({type(e).__name__}: {e})')
if missing:
    print('weights not yet in TORCH_HOME (downloaded when a model needs them if this machine has internet access, otherwise the '
          'model stops). To download them beforehand:\n  mkdir -p ' + CKPT_DIR
          + ''.join(f'\n  wget -O {f} {u}' for f, u in missing.items()))
elif todo:
    print('all weights of the models to run are in', CKPT_DIR)
if 'olmoearth' in todo:
    try:
        import olmoearth_pretrain_minimal
        print('olmoearth_pretrain_minimal', getattr(olmoearth_pretrain_minimal, '__version__', ''), 'from',
              os.path.dirname(olmoearth_pretrain_minimal.__file__))
    except ImportError:
        print('WARNING: olmoearth_pretrain_minimal is not installed. ' + OLMOEARTH_HINT)
````

````python
# the temporal transformer on random NDVI series (4 x 3 x 3 cells), in float64 on the CPU: how its 6 bands change when
# every series is shifted by 1 ... 22 composites (cyclically) and when the composites are put in a random order. The
# check uses large random weights (sd 0.5), so that the attention is far from uniform (at the start of training the
# weights are small and every setting is still close to the mean); the structure, not the starting point, is checked.
torch.manual_seed(0)
nd_test = 0.2 + 0.6 * torch.rand(4, N_COMPOSITES, 3, 3, dtype=torch.float64)
order = torch.randperm(N_COMPOSITES)
rows_check = []
for mode in POSITIONS:
    enc = TemporalEncoder(mode).double().eval()
    n_params = sum(p.numel() for p in enc.parameters())
    with torch.no_grad():
        start_sd = float(enc(nd_test).std())                  # the bands at the start of training
        for p in enc.parameters():
            p.normal_(0, 0.5)
        ref = enc(nd_test)
        shifted = max(float((enc(torch.roll(nd_test, s, dims=1)) - ref).abs().max()) for s in range(1, N_COMPOSITES))
        shuffled = float((enc(nd_test[:, order]) - ref).abs().max())
    rows_check.append({'TEMPORAL_POSITION': mode, 'parameters': n_params, 'sd of the bands at the start': start_sd,
                       'bands: sd': float(ref.std()), 'max change, series shifted': shifted,
                       'max change, composites reordered': shuffled})
check = pd.DataFrame(rows_check).set_index('TEMPORAL_POSITION')
tol = 1e-9 * check['bands: sd']
assert (check.loc[['cyclic', 'none'], 'max change, series shifted'] < tol[['cyclic', 'none']]).all(), check
assert check.at['none', 'max change, composites reordered'] < tol['none'], check
assert check.at['cyclic', 'max change, composites reordered'] > 1e-3 * check.at['cyclic', 'bands: sd'], check
assert check.at['absolute', 'max change, series shifted'] > 1e-3 * check.at['absolute', 'bands: sd'], check
print('cyclic: invariant to seasonal shifts, not to reordering | none: invariant to both | absolute: to neither')
print(f'this run: TEMPORAL_POSITION = {TEMPORAL_POSITION!r}, {check.at[TEMPORAL_POSITION, "parameters"]} parameters;'
      f' one batch of {BATCH_SIZE} images = {BATCH_SIZE * CELLS * CELLS} cells x {N_COMPOSITES} tokens')

if DEVICE.type == 'cuda':                                     # time of the transformer for one training batch
    enc = TemporalEncoder().to(DEVICE)
    nd_batch = 0.2 + 0.6 * torch.rand(BATCH_SIZE, N_COMPOSITES, CELLS, CELLS, device=DEVICE)
    for _ in range(3):                                        # the first passes include CUDA's start-up
        torch.cuda.synchronize()
        t0 = time.time()
        with torch.autocast(device_type='cuda', dtype=torch.bfloat16, enabled=AMP):
            out = enc(nd_batch)
        out.float().square().mean().backward()
        torch.cuda.synchronize()
    print(f'temporal transformer, forward and backward of one batch: {1000 * (time.time() - t0):.0f} ms, peak GPU '
          f'memory {torch.cuda.max_memory_allocated() / 1e9:.1f} GB')
    del enc, nd_batch, out
    torch.cuda.empty_cache()
check.round(12)
````

````python
class Inputs:
    """The inputs of the models for the clusters at the rows `rows` of `clusters`: the 7 image bands, normalised as in
    the IWI study (normalise()) and in DEVICE memory when they take less than GPU_DATA_GB (otherwise every batch is read from
    the caches, which is slower); the 23 composites of every cell (NDVI), in DEVICE memory; and `loc`, the SatCLIP
    values of their coordinates (set in section 5 when a model needs them)."""

    def __init__(self, rows, load_images=True, load_ndvi=True):
        self.rows = np.asarray(rows)
        self.x, self.nd, self.loc = None, None, None
        if load_images:
            gb = len(self.rows) * len(IMAGE) * PATCH * PATCH * 2 / 1e9
            if gb < GPU_DATA_GB:
                self.x = torch.empty((len(self.rows), len(IMAGE), PATCH, PATCH), dtype=torch.float16, device=DEVICE)
                for s in tqdm(range(0, len(self.rows), 256), desc=f'loading {len(IMAGE)} bands, {gb:.1f} GB',
                              leave=False):
                    self.x[s:s + 256] = torch.from_numpy(read_input(self.rows[s:s + 256])).to(DEVICE)
            print(f'{len(IMAGE)} image bands of {len(self.rows)} clusters, {gb:.1f} GB: ' + (
                  'in GPU memory' if self.x is not None and DEVICE.type == 'cuda' else 'in memory' if self.x is not None
                  else f'more than GPU_DATA_GB = {GPU_DATA_GB}, so every batch is read from the caches'))
        if load_ndvi:
            gb = len(self.rows) * N_COMPOSITES * CELLS * CELLS * 2 / 1e9
            self.nd = torch.empty((len(self.rows), N_COMPOSITES, CELLS, CELLS), dtype=torch.float16, device=DEVICE)
            for s in range(0, len(self.rows), 4096):
                self.nd[s:s + 4096] = torch.from_numpy(read_ndvi(self.rows[s:s + 4096])).to(DEVICE)
            print(f'{N_COMPOSITES} NDVI composites of {len(self.rows)} clusters ({CELLS} x {CELLS} cells), '
                  f'{gb:.2f} GB: ' + ('in GPU memory' if DEVICE.type == 'cuda' else 'in memory'))

    def images(self, idx):
        """Normalised float32 image bands (channels last) of the clusters at positions idx (a 1-D tensor)."""
        if self.x is not None:
            x = self.x[idx.to(self.x.device)]
        else:
            x = torch.from_numpy(read_input(self.rows[idx.cpu().numpy()])).to(DEVICE, non_blocking=True)
        idx = idx.to(DEVICE)
        x = normalise(x.float())
        return x.contiguous(memory_format=torch.channels_last)

    def ndvi(self, idx):
        """NDVI composites (float32, (n, 23, CELLS, CELLS)) of the clusters at positions idx (a 1-D tensor)."""
        return self.nd[idx.to(self.nd.device)].to(DEVICE).float()

    def batch(self, idx, needs):
        """(image bands or None, NDVI composites or None, SatCLIP values or None) of the clusters at positions idx, for
        a model with inputs `needs`."""
        x = self.images(idx) if 'image' in needs else None
        nd = self.ndvi(idx) if 'ndvi' in needs else None
        loc = self.loc[idx.to(self.loc.device)] if 'location' in needs else None
        return x, nd, loc


def r2_score(y, p):
    y, p = np.asarray(y, np.float64), np.asarray(p, np.float64)
    return float(1 - ((y - p) ** 2).sum() / ((y - y.mean()) ** 2).sum())


def autocast():
    return torch.autocast(device_type=DEVICE.type, dtype=torch.bfloat16, enabled=AMP and DEVICE.type == 'cuda')


NO_DECAY = ('relative_position', 'cpb_mlp', 'logit_scale', 'month_embed', 'channel_embed')   # Swin v2, OlmoEarth


def make_optimizer(model, recipe):
    """adam: Adam with LR for all weights and the Keras epsilon (05-07). adamw: AdamW as in 08, LR_PRETRAINED for the
    pretrained weights (weight decay on their matrices, not on biases, norms, embeddings and position biases) and
    LR_NEW without weight decay for the new ones (Regressor.is_new). Every group keeps its base_lr for the schedule."""
    params = [(n, p) for n, p in model.named_parameters() if p.requires_grad]
    if recipe == 'adam':
        opt = torch.optim.Adam([{'params': [p for _, p in params], 'lr': LR, 'name': 'all'}], eps=1e-7)
    else:
        groups = {'decay': [], 'no_decay': [], 'new': []}
        for name, p in params:
            if model.is_new(name):
                groups['new'].append(p)
            elif p.ndim >= 2 and not name.endswith(('pos_embed', 'cls_token')) and not any(t in name for t in NO_DECAY):
                groups['decay'].append(p)
            else:
                groups['no_decay'].append(p)
        spec = [{'params': groups['decay'], 'lr': LR_PRETRAINED, 'weight_decay': WEIGHT_DECAY, 'name': 'pretrained'},
                {'params': groups['no_decay'], 'lr': LR_PRETRAINED, 'weight_decay': 0.0,
                 'name': 'pretrained_no_decay'},
                {'params': groups['new'], 'lr': LR_NEW, 'weight_decay': 0.0, 'name': 'new'}]
        opt = torch.optim.AdamW([g for g in spec if g['params']])
    for g in opt.param_groups:
        g['base_lr'] = g['lr']
    return opt


def lr_factor(t):
    """adamw: learning-rate factor at epoch t (fractional): linear warm-up over WARMUP_EPOCHS, cosine decay to 0."""
    if t < WARMUP_EPOCHS:
        return t / WARMUP_EPOCHS
    return 0.5 * (1 + math.cos(math.pi * min(1.0, (t - WARMUP_EPOCHS) / max(EPOCHS - WARMUP_EPOCHS, 1))))


def predict(model, data, idx, needs):
    """Predictions (bfloat16 autocast on the GPU) of the clusters at positions idx, in batches of 2 x BATCH_SIZE."""
    model.eval()
    out = []
    batch = 2 * BATCH_SIZE
    with torch.no_grad(), autocast():
        for s in range(0, len(idx), batch):
            out.append(model(*data.batch(idx[s:s + batch], needs)).float())
    return torch.cat(out)


def extract(model, data, idx, needs):
    """Features (n, n_features) and predictions (n,) of the clusters at positions idx, in float32 (no autocast), in
    batches of BATCH_SIZE (float32 needs more memory than training in bfloat16)."""
    model.eval()
    batch = BATCH_SIZE
    feats, preds = [], []
    with torch.no_grad():
        for s in range(0, len(idx), batch):
            f = model.features(*data.batch(idx[s:s + batch], needs)).float()
            feats.append(f.cpu())
            preds.append(model.head(f).squeeze(1).cpu())
    return torch.cat(feats).numpy(), torch.cat(preds).numpy()


def save(obj, path):
    torch.save(obj, path + '.tmp')
    os.replace(path + '.tmp', path)


def write_csv(frame, path, **kw):
    """Write a CSV file under a temporary name, then rename it: nobody (another run) reads half a file."""
    tmp = f'{path}.{uuid.uuid4().hex[:8]}.tmp'
    frame.to_csv(tmp, index=False, **kw)
    os.replace(tmp, path)


def check_start(from_scratch, where, config=None):
    """A run folder holds one kind of run: stop instead of mixing runs from pretrained and from random weights, or
    runs with different temporal transformers (config: the TEMPORAL_CONFIG of the run found, None: not checked)."""
    if bool(from_scratch) != TRAIN_FROM_SCRATCH:
        kind = {True: 'from random weights', False: 'from the pretrained weights'}
        raise RuntimeError(f'{where} holds a run {kind[bool(from_scratch)]}, but this run starts '
                           f'{kind[TRAIN_FROM_SCRATCH]}: give it another RUN_ROOT')
    if config is not None and dict(config) != TEMPORAL_CONFIG:
        diff = {k: (config.get(k), TEMPORAL_CONFIG.get(k)) for k in sorted(set(config) | set(TEMPORAL_CONFIG))
                if config.get(k) != TEMPORAL_CONFIG.get(k)}
        raise RuntimeError(f'{where} holds a run with another temporal transformer (setting: (that run, Settings) = '
                           f'{diff}): change Settings back or give this run another RUN_ROOT')
````

````python
META = ['cluster_id', 'country', 'year', 'rural', 'lat', 'lon', 'nl_sensor', 'fold']   # columns of predictions.csv


def split(y, folds, test_fold):
    """Positions of the training, validation and test clusters (with a target) of a test fold, its validation fold and
    training folds."""
    k = FOLDS.index(test_fold)
    val_fold = FOLDS[(k - 1) % len(FOLDS)]
    train_folds = [f for f in FOLDS if f not in (test_fold, val_fold)]
    has_y = np.isfinite(y)
    return (np.flatnonzero(has_y & np.isin(folds, train_folds)), np.flatnonzero(has_y & (folds == val_fold)),
            np.flatnonzero(has_y & (folds == test_fold)), val_fold, train_folds)


def write_outputs(key, run_dir, data, y, folds, test_fold, pred, feats=None, extra=None):
    """predictions.csv (every cluster of data, all sets), features.csv (SAVE_FEATURES and feats), test_predictions.csv
    and result.json of a finished model of one test fold. Returns the result."""
    tr, va, te, val_fold, train_folds = split(y, folds, test_fold)
    role = np.select([np.isin(folds, train_folds), folds == val_fold, folds == test_fold], ['train', 'val', 'test'], '')
    n_set = {r: int((role == r).sum()) for r in ('train', 'val', 'test', '')}      # rows per set (incl. no target)
    out = pd.concat([clusters.iloc[data.rows][[c for c in META if c in clusters]].reset_index(drop=True),
                     pd.DataFrame({'role': role, TARGET: y, f'{TARGET}_pred': pred})], axis=1)
    write_csv(out.sort_values('cluster_id'), os.path.join(run_dir, 'predictions.csv'), float_format='%.7g')
    if feats is not None and SAVE_FEATURES:
        cols = pd.DataFrame(feats, columns=[f'lt_{key}_feat_{j}' for j in range(feats.shape[1])])
        write_csv(pd.concat([out, cols], axis=1).sort_values('cluster_id'), os.path.join(run_dir, 'features.csv'),
                  float_format='%.7g')
    p_te, y_te = pred[te], y[te]
    write_csv(clusters.iloc[data.rows[te]][['cluster_id', 'country', 'year', 'nl_sensor']].assign(y_true=y_te,
                                                                                                  y_pred=p_te),
              os.path.join(run_dir, 'test_predictions.csv'))
    name, _, recipe, needs = MODELS[key]
    result = {'key': key, 'model': name, 'start': start_of(key), 'recipe': recipe, 'inputs': '+'.join(needs),
              'test_fold': test_fold, 'val_fold': val_fold, 'train_folds': ','.join(train_folds),
              'keras_name': KERAS_NAME[test_fold], 'sensors': ','.join(NL_SENSORS), 'target': TARGET, 'n_train': len(tr),
              'n_val': len(va), 'n_test': len(te), 'n_clusters': len(out),
              'n_features': None if feats is None else int(feats.shape[1]), 'val_r2': r2_score(y[va], pred[va]),
              'test_r2': r2_score(y_te, p_te), 'test_rmse': float(np.sqrt(((p_te - y_te) ** 2).mean())),
              'test_mae': float(np.abs(p_te - y_te).mean()), 'predictions_train': n_set['train'],
              'predictions_val': n_set['val'], 'predictions_test': n_set['test'], 'from_scratch': TRAIN_FROM_SCRATCH,
              'ndvi': ndvi_route(key), 'temporal_position': TEMPORAL_POSITION, 'temporal': TEMPORAL_CONFIG,
              **(extra or {})}
    result_file = os.path.join(run_dir, 'result.json')
    with open(result_file + '.tmp', 'w') as f:
        json.dump(result, f, indent=1)
    os.replace(result_file + '.tmp', result_file)
    print(f"{key} test {test_fold} ({KERAS_NAME[test_fold]}): test R2 {result['test_r2']:.3f}, RMSE "
          f"{result['test_rmse']:.2f}" + (f" (best epoch {result['best_epoch']})" if 'best_epoch' in result else '')
          + f"\n    predictions of all sets -> {run_dir}: train {n_set['train']} ({', '.join(train_folds)}), val "
          f"{n_set['val']} ({val_fold}), test {n_set['test']} ({test_fold})" + (f", no fold {n_set['']}"
                                                                                 if n_set[''] else ''))
    return result


def finished(key, f):
    """A model of one test fold is done: result.json, and features.csv for a network with SAVE_FEATURES."""
    run_dir = os.path.join(RUN_ROOT, key, f'test_{f}')
    need = ['result.json'] + (['features.csv'] if SAVE_FEATURES and MODELS[key][2] != 'ridge' else [])
    return all(os.path.exists(os.path.join(run_dir, n)) for n in need)


def run_fold(key, data, y, folds, test_fold):
    """Train (or continue) the network `key` of one test fold, then write the predictions (and features) of every
    cluster, the test predictions and result.json. y: the target at the positions of data.rows (NaN: none); folds:
    their folds of 03 ('': none). Returns the test metrics (also in result.json)."""
    if MODELS[key][2] == 'ridge':
        return run_mosaiks(key, data, y, folds, test_fold)
    _, _, recipe, needs = MODELS[key]
    run_dir = os.path.join(RUN_ROOT, key, f'test_{test_fold}')
    result_file = os.path.join(run_dir, 'result.json')
    if os.path.exists(result_file):
        with open(result_file) as f:
            done = json.load(f)
        check_start(done.get('from_scratch', False), run_dir, done.get('temporal'))
        if finished(key, test_fold):
            return done
    os.makedirs(run_dir, exist_ok=True)
    k = FOLDS.index(test_fold)
    tr, va, te, val_fold, train_folds = (torch.from_numpy(v).to(DEVICE) if isinstance(v, np.ndarray) else v
                                         for v in split(y, folds, test_fold))
    y_dev = torch.tensor(np.nan_to_num(y), dtype=torch.float32, device=DEVICE)
    assert len(tr) >= BATCH_SIZE, f'{key} test {test_fold}: {len(tr)} training clusters, fewer than BATCH_SIZE'

    torch.manual_seed(SEED + k)
    model = build_model(key).to(DEVICE)
    if 'image' in needs:
        model = model.to(memory_format=torch.channels_last)
    if 'location' in needs:
        model.set_location_scale(data.loc[tr])
    with torch.no_grad():
        model.head.bias.fill_(float(y_dev[tr].mean()))
    opt = make_optimizer(model, recipe)
    gen = torch.Generator(device=DEVICE)
    gen.manual_seed(SEED * 100 + k)
    start, best, best_epoch = 0, float('inf'), -1
    last_file, best_file, hist_file = (os.path.join(run_dir, f) for f in ('last.pt', 'best.pt', 'history.csv'))
    if os.path.exists(result_file):                     # finished before: only the features are written again
        assert os.path.exists(best_file), f'{best_file} is missing: delete {result_file} to train this model again'
        start = EPOCHS
    elif os.path.exists(last_file):                     # continue a run that was stopped
        st = torch.load(last_file, map_location=DEVICE)
        check_start(st.get('from_scratch', False), run_dir, st.get('temporal'))
        model.load_state_dict(st['model'])
        opt.load_state_dict(st['opt'])
        gen.set_state(st['gen'].cpu())
        start, best, best_epoch = st['epoch'] + 1, st['best'], st['best_epoch']
        if os.path.exists(hist_file):                   # epochs after the saved state are trained again
            with open(hist_file) as fh:
                kept = fh.readlines()[:start + 1]       # the header and epochs 1 ... start, as they were written
            with open(hist_file, 'w') as fh:
                fh.writelines(kept)
        print(f'{key} test {test_fold}: continuing at epoch {start + 1}')
    elif os.path.exists(hist_file):
        os.remove(hist_file)                            # history of a run that stopped before saving its state
    n_temporal = sum(p.numel() for p in model.temporal.parameters()) if model.temporal is not None else 0
    if start == 0:
        print(f'{key} test {test_fold}: {sum(p.numel() for p in model.parameters()) / 1e6:.1f} M parameters ('
              f'{n_temporal / 1e6:.3f} M in the temporal transformer), {model.n_features} features, {recipe}, start: '
              f'{start_of(key)}')

    n_steps = max(len(tr) // BATCH_SIZE, 1)
    for epoch in range(start, EPOCHS):
        t0 = time.time()
        model.train()
        perm = tr[torch.randperm(len(tr), generator=gen, device=DEVICE)]
        sq_sum = torch.zeros((), device=DEVICE)
        n_seen = 0
        for j, s in enumerate(range(0, len(perm) - BATCH_SIZE + 1, BATCH_SIZE)):    # last incomplete batch dropped
            if recipe == 'adamw':                       # warm-up and cosine decay, every step
                for g in opt.param_groups:
                    g['lr'] = g['base_lr'] * lr_factor(epoch + (j + 1) / n_steps)
            b = perm[s:s + BATCH_SIZE]
            x, nd, loc = data.batch(b, needs)
            with autocast():
                pred = model(x, nd, loc)
            mse = F.mse_loss(pred.float(), y_dev[b])
            loss = mse + L2_HEAD * model.head.weight.pow(2).sum() if L2_HEAD else mse
            opt.zero_grad(set_to_none=True)
            loss.backward()
            opt.step()
            sq_sum += mse.detach() * len(b)
            n_seen += len(b)
        reg = L2_HEAD * float(model.head.weight.detach().pow(2).sum())
        p_va = predict(model, data, va, needs).cpu().numpy()
        y_va = y[va.cpu().numpy()]
        val_mse = float(((p_va - y_va) ** 2).mean())
        val_loss = val_mse + reg
        if val_loss < best:
            best, best_epoch = val_loss, epoch
            save({'model': model.state_dict(), 'epoch': epoch, 'from_scratch': TRAIN_FROM_SCRATCH,
                  'temporal': TEMPORAL_CONFIG}, best_file)
        lrs = {f"lr_{g['name']}": g['lr'] for g in opt.param_groups}
        rec = {'epoch': epoch + 1, 'train_mse': float(sq_sum) / max(n_seen, 1), 'l2_term': reg, 'val_mse': val_mse,
               'val_loss': val_loss, 'val_r2': r2_score(y_va, p_va), **lrs, 'seconds': round(time.time() - t0, 1)}
        pd.DataFrame([rec]).to_csv(hist_file, mode='a', header=not os.path.exists(hist_file), index=False)
        save({'model': model.state_dict(), 'opt': opt.state_dict(), 'gen': gen.get_state(), 'epoch': epoch,
              'best': best, 'best_epoch': best_epoch, 'from_scratch': TRAIN_FROM_SCRATCH, 'temporal': TEMPORAL_CONFIG},
             last_file)
        if epoch == start or (epoch + 1) % 10 == 0 or epoch + 1 == EPOCHS:
            print(f"{key} test {test_fold} epoch {epoch + 1}/{EPOCHS}: train mse {rec['train_mse']:.2f}, val mse "
                  f"{val_mse:.2f}, val R2 {rec['val_r2']:.3f}, best epoch {best_epoch + 1}, {rec['seconds']} s")

    # the best weights: features and prediction of every cluster (all sets), test metrics
    st = torch.load(best_file, map_location=DEVICE)
    check_start(st.get('from_scratch', False), run_dir, st.get('temporal'))
    model.load_state_dict(st['model'])
    feats, pred = extract(model, data, torch.arange(len(data.rows), device=DEVICE), needs)
    result = write_outputs(key, run_dir, data, y, folds, test_fold, pred, feats,
                           {'best_epoch': st['epoch'] + 1, 'epochs': EPOCHS, 'temporal_parameters': n_temporal})
    if os.path.exists(last_file):
        os.remove(last_file)                            # only needed to continue a stopped run
    return result
````

````python
def mosaiks_images(data, idx):
    """The 30 MOSAIKS bands (MOSAIKS_INPUT order) of the clusters at positions idx (a 1-D tensor): the 7 normalised
    image bands and the 23 NDVI composites, every pixel with the values of its cell (float32, (n, 30, 224, 224))."""
    x = data.images(idx)
    nd = F.interpolate(data.ndvi(idx), size=x.shape[-2:], mode='nearest')
    return torch.cat([x, nd.to(x.dtype)], dim=1)


def band_scaling(data, idx, n_max=1000):
    """Mean and standard deviation of every MOSAIKS band over the pixels of (up to n_max of) the clusters at positions
    idx (a 1-D tensor), float32 tensors on DEVICE."""
    pick = idx[torch.randperm(len(idx), generator=torch.Generator().manual_seed(SEED))[:n_max].to(idx.device)]
    s1 = torch.zeros(len(MOSAIKS_INPUT), dtype=torch.float64, device=DEVICE)
    s2 = torch.zeros_like(s1)
    for s in range(0, len(pick), 64):
        x = mosaiks_images(data, pick[s:s + 64]).double()
        s1 += x.sum(dim=(0, 2, 3))
        s2 += (x ** 2).sum(dim=(0, 2, 3))
    n = len(pick) * PATCH * PATCH
    mean = s1 / n
    return mean.float(), (s2 / n - mean ** 2).clamp(min=1e-12).sqrt().float()


class ScaledImages:
    """The standardised MOSAIKS bands of the clusters at positions idx of `data` as a dataset for torchgeo's RCF:
    item i = {'image': tensor (30, 224, 224)}."""

    def __init__(self, data, idx, mean, std):
        self.data, self.idx = data, idx
        self.mean, self.std = mean.view(1, -1, 1, 1), std.view(1, -1, 1, 1)

    def __len__(self):
        return len(self.idx)

    def __getitem__(self, i):
        x = (mosaiks_images(self.data, self.idx[int(i):int(i) + 1]) - self.mean) / self.std
        return {'image': x[0].float().cpu()}


def ridge(X, y, alphas):
    """Ridge regression of y on X with intercept (not penalised) for every alpha, the features standardised with the
    mean and standard deviation of X (constant features: scale 1), as in 09 of the IWI study; float64 tensors. Returns the coefficients
    on the raw features (len(alphas), n_features) and the intercepts (len(alphas),)."""
    mean, std = X.mean(dim=0), X.std(dim=0, unbiased=False)
    std = torch.where(std < 1e-12, torch.ones_like(std), std)
    Z = (X - mean) / std
    y_mean = y.mean()
    s, V = torch.linalg.eigh(Z.T @ Z)
    s = s.clamp(min=0)
    Vb = V.T @ (Z.T @ (y - y_mean))
    coef = torch.stack([V @ (Vb / (s + a)) for a in alphas]) / std
    return coef, y_mean - coef @ mean


def run_mosaiks(key, data, y, folds, test_fold):
    """MOSAIKS of one test fold: band scaling and RCF filters from the training clusters, the features of every
    cluster, ridge (alpha on the validation fold, then fitted on the training and validation folds), outputs."""
    run_dir = os.path.join(RUN_ROOT, key, f'test_{test_fold}')
    result_file = os.path.join(run_dir, 'result.json')
    if os.path.exists(result_file):
        with open(result_file) as f:
            done = json.load(f)
        check_start(done.get('from_scratch', False), run_dir, done.get('temporal'))
        return done
    os.makedirs(run_dir, exist_ok=True)
    from torchgeo.models import RCF
    t0 = time.time()
    k = FOLDS.index(test_fold)
    tr, va, te, _, _ = split(y, folds, test_fold)
    idx_tr = torch.from_numpy(tr).to(DEVICE)
    mean, std = band_scaling(data, idx_tr)
    torch.manual_seed(SEED + k)
    rcf = RCF(in_channels=len(MOSAIKS_INPUT), features=MOSAIKS_FEATURES, kernel_size=MOSAIKS_KERNEL, bias=MOSAIKS_BIAS,
              seed=SEED + k + 1, mode='empirical', dataset=ScaledImages(data, idx_tr, mean, std)).to(DEVICE)
    X = torch.empty((len(data.rows), MOSAIKS_FEATURES), dtype=torch.float64, device=DEVICE)
    with torch.no_grad():
        for s in tqdm(range(0, len(data.rows), MOSAIKS_BATCH), desc=f'MOSAIKS features, test {test_fold}', leave=False):
            idx = torch.arange(s, min(s + MOSAIKS_BATCH, len(data.rows)), device=DEVICE)
            x = (mosaiks_images(data, idx) - mean.view(1, -1, 1, 1)) / std.view(1, -1, 1, 1)
            X[s:s + len(idx)] = rcf(x.contiguous()).double()
    t_feat = time.time() - t0
    yd = torch.tensor(np.nan_to_num(y), dtype=torch.float64, device=DEVICE)
    tr_d, va_d = torch.from_numpy(tr).to(DEVICE), torch.from_numpy(va).to(DEVICE)
    coef, icpt = ridge(X[tr_d], yd[tr_d], ALPHAS)                        # alpha: fitted on train, tested on val
    p_va = X[va_d] @ coef.T + icpt
    a = int(torch.argmin(((p_va - yd[va_d, None]) ** 2).mean(dim=0)))
    alpha, val_r2 = float(ALPHAS[a]), r2_score(y[va], p_va[:, a].cpu().numpy())
    fit = torch.from_numpy(np.concatenate([tr, va])).to(DEVICE)
    coef, icpt = ridge(X[fit], yd[fit], [alpha])                         # final: train + val
    pred = (X @ coef[0] + icpt[0]).cpu().numpy()
    save({'filters': rcf.weights.cpu(), 'biases': rcf.biases.cpu(), 'band_mean': mean.cpu(), 'band_std': std.cpu(),
          'alpha': alpha, 'coef': coef[0].cpu(), 'intercept': float(icpt[0])}, os.path.join(run_dir, 'mosaiks.pt'))
    edge = alpha in (ALPHAS[0], ALPHAS[-1])
    if edge:
        print(f'NOTE: {key} test {test_fold}: alpha {alpha:g} is at the edge of ALPHAS; widen ALPHAS')
    print(f'{key} test {test_fold}: {MOSAIKS_FEATURES} features of {len(data.rows)} clusters in {t_feat:.0f} s, '
          f'alpha {alpha:g} (validation R2 {val_r2:.3f})')
    return write_outputs(key, run_dir, data, y, folds, test_fold, pred, None,
                         {'n_features': MOSAIKS_FEATURES, 'alpha': alpha, 'alpha_at_grid_edge': edge,
                          'val_r2': val_r2, 'seconds': round(time.time() - t0, 1)})
````

# 5. Train

The models train on the clusters with complete data, a target value and a fold of 03, and predict all clusters with complete data. The 7 image bands and the 23 composites of every cell are loaded once for all models. Models and test folds come from `RUN_MODELS` and `RUN_FOLDS`, the position setting of the temporal transformer from `TEMPORAL_POSITION`.

````python
rows = np.flatnonzero(complete)                                # rows of `clusters` that get predictions
y = clusters[TARGET].to_numpy(dtype=np.float64)[rows]          # NaN: no target, not trained on
folds = clusters['fold'].fillna('').astype(str).to_numpy()[rows]
trainable = np.isfinite(y) & np.isin(folds, FOLDS)
print(f'{len(rows)} clusters with complete data, {trainable.sum()} of them with {TARGET} and a fold:',
      pd.Series(folds[trainable]).value_counts().sort_index().to_dict())

runs = {key: [f for f in RUN_FOLDS if not finished(key, f)] for key in todo}
needs = {n for key, missing in runs.items() if missing for n in MODELS[key][3]}
data = Inputs(rows, load_images='image' in needs, load_ndvi='ndvi' in needs) if needs else None
if 'location' in needs:
    data.loc = satclip_embeddings(rows)
    print(f'SatCLIP values of {len(rows)} clusters: {tuple(data.loc.shape)}')
if needs and DEVICE.type != 'cuda':
    print('WARNING: no GPU, training will be very slow')
for key in todo:
    if not runs[key]:
        print(f'{key}: test folds {RUN_FOLDS} already done')
        continue
    print(f'\n=== {key}: {MODELS[key][0]}, start: {start_of(key)}, NDVI: {ndvi_route(key)}, test folds {runs[key]}')
    for f in runs[key]:
        run_fold(key, data, y, folds, f)
    if DEVICE.type == 'cuda':
        torch.cuda.empty_cache()
if not todo:
    print('LT_MODELS=none: no training, only the results')
````

# 6. Results

For every model: the mean and standard deviation of the test R² over the test folds that are done, and the R², RMSE and MAE (in units of the target: NormFunc standard deviations of the period) of all test predictions pooled, and the pooled R² of each period on its own (`pooled R2 DMSP`, `pooled R2 VIIRS`: the R² in sensor units). Next to them, the same model in the runs it is meant to be compared with, all with the same `NL_SENSORS` and start:

* **11, the 6 wavelet bands** (`WAVELET_RUNS`): its mean test R², and `15 − 11`, the test R² of 15 minus that of 11 for every test fold that both have done with the same training, validation and test clusters (mean, standard deviation and number of folds). Above 0: the temporal transformer on the 23 raw composites does better than the wavelet bands.
* **12, Landsat only** (`LANDSAT_RUNS`): the same columns, `15 − 12` being what the NDVI series brings through the temporal transformer. 12 against 11 gives the same for the wavelet bands.
* **15 with the other position settings** (`POSITION_RUNS`, when they have run): `15 − 15 none` and `15 − 15 absolute` (or `cyclic`, depending on the setting of this run), the difference made by the way the transformer knows the time.

````python
def scores(p):
    err = p['y_pred'] - p['y_true']
    return {'clusters': len(p), 'R2': r2_score(p['y_true'], p['y_pred']) if len(p) > 1 else np.nan,
            'RMSE': float(np.sqrt((err ** 2).mean())), 'MAE': float(err.abs().mean())}


def fold_results(key, run_root=None, check=True):
    """result.json of every finished test fold of a model, and its pooled test predictions (or None)."""
    found, preds = [], []
    for f in FOLDS:
        run_dir = os.path.join(run_root or RUN_ROOT, key, f'test_{f}')
        if os.path.exists(os.path.join(run_dir, 'result.json')):
            with open(os.path.join(run_dir, 'result.json')) as fh:
                found.append(json.load(fh))
            if check:                                  # a run folder holds one kind of run
                check_start(found[-1].get('from_scratch', False), run_dir, found[-1].get('temporal'))
            preds.append(pd.read_csv(os.path.join(run_dir, 'test_predictions.csv')))
    return found, (pd.concat(preds, ignore_index=True) if preds else None)


def reference_results():
    """[(label, mean test R2)] of the comparable runs of 05 and 07 of this study that are done."""
    refs = []
    for start in REF_STARTS:
        f = os.path.join(WORK_DIR, 'landsat_resnet18_runs', f'{PERIODS}_{start}', 'landsat_resnet18_results.csv')
        if os.path.exists(f):
            t = pd.read_csv(f).set_index('key')
            if 'landsat_wavelet6' in t.index and t.at['landsat_wavelet6', 'folds done'] > 0:
                refs.append((f'Landsat + wavelets, ResNet-18 (07, {start})',
                             float(t.at['landsat_wavelet6', 'test R2 mean'])))
                break
    if PERIODS == 'VIIRS':
        r = [x['test_r2'] for x in fold_results('base_wavelet6', T2_RESNET18, check=False)[0]]
        if r:
            refs.append((f'base_wavelet6, ResNet-18 (05, {len(r)} folds)', float(np.mean(r))))
    return refs


per_fold, rows_out, oof = [], [], []
for key, (name, _, recipe, needs) in MODELS.items():
    found, p = fold_results(key)
    per_fold += found
    r = pd.Series([x['test_r2'] for x in found], dtype=float)
    out = {'key': key, 'model': name, 'start': start_of(key), 'training': recipe, 'folds done': len(r),
           'test R2 mean': r.mean() if len(r) else np.nan, 'test R2 sd': r.std() if len(r) > 1 else np.nan,
           'test R2 min': r.min() if len(r) else np.nan, 'test R2 max': r.max() if len(r) else np.nan}
    if p is not None:
        sc = scores(p)
        out.update({'pooled R2': sc['R2'], 'pooled RMSE': sc['RMSE'], 'pooled MAE': sc['MAE']})
        for s, ps in p.groupby('nl_sensor'):
            out[f'pooled R2 {s}'] = scores(ps)['R2']
        oof.append(p.set_index('cluster_id')['y_pred'].rename(key))
    rows_out.append(out)
# the same models in 11 (wavelet bands), 12 (Landsat only) and in 15 with the other position settings: their mean test R2,
# and the test R2 of this run minus theirs, paired by test fold (folds with the same numbers of training, validation
# and test clusters in both)
COMPARE = {'11': ('wavelets', WAVELET_RUNS), '12': ('Landsat only', LANDSAT_RUNS)}
COMPARE.update({f'15 {p}': (f'position {p}', r) for p, r in POSITION_RUNS.items()})
n_compare = dict.fromkeys(COMPARE, 0)
for out in rows_out:
    mine = {x['test_fold']: x for x in per_fold if x['key'] == out['key']}
    for label, (what, root) in COMPARE.items():
        theirs = {x['test_fold']: x for x in fold_results(out['key'], root, check=False)[0]}
        n_compare[label] += bool(theirs)
        out[f'R2 {label} ({what})'] = np.mean([x['test_r2'] for x in theirs.values()]) if theirs else np.nan
        gain = [mine[f]['test_r2'] - theirs[f]['test_r2'] for f in mine if f in theirs and
                all(theirs[f][n] == mine[f][n] for n in ('n_train', 'n_val', 'n_test'))]
        out[f'15 - {label}'] = np.mean(gain) if gain else np.nan
        out[f'15 - {label} sd'] = np.std(gain, ddof=1) if len(gain) > 1 else np.nan
        out[f'paired folds {label}'] = len(gain)
CMP_COLS = [c for label, (what, _) in COMPARE.items()
            for c in (f'R2 {label} ({what})', f'15 - {label}', f'15 - {label} sd', f'paired folds {label}')]
RES_COLS = ['key', 'model', 'start', 'training', 'folds done', 'test R2 mean', 'test R2 sd', 'test R2 min',
            'test R2 max', 'pooled R2', 'pooled RMSE', 'pooled MAE'] + [f'pooled R2 {s}' for s in NL_SENSORS] + CMP_COLS
results = pd.DataFrame(rows_out, columns=RES_COLS)
FOLD_COLS = ['key', 'test_fold', 'val_fold', 'train_folds', 'keras_name', 'n_train', 'n_val', 'n_test', 'best_epoch',
             'alpha', 'val_r2', 'test_r2', 'test_rmse', 'test_mae']
per_fold = pd.DataFrame(per_fold, columns=FOLD_COLS + ['model', 'start', 'recipe', 'n_features', 'n_clusters',
                                                    'sensors', 'epochs', 'ndvi', 'temporal_position',
                                                    'temporal_parameters'])
os.makedirs(RUN_ROOT, exist_ok=True)
write_csv(results, os.path.join(RUN_ROOT, 'lt_results.csv'))
write_csv(per_fold, os.path.join(RUN_ROOT, 'lt_per_fold.csv'))

OOF = None
if oof:
    base = clusters.iloc[rows][[c for c in ['cluster_id', 'country', 'year', 'rural', 'nl_sensor', 'fold', TARGET]
                                if c in clusters]].set_index('cluster_id')
    OOF = base.join(pd.concat(oof, axis=1), how='inner').reset_index()
    write_csv(OOF.sort_values('cluster_id'), os.path.join(RUN_ROOT, 'oof_predictions.csv'), float_format='%.6g')

REFS = reference_results()
print(f'Landsat + temporal transformer ({TEMPORAL_POSITION}) on the {N_COMPOSITES} NDVI composites -> {TARGET} '
      f'({" + ".join(NL_SENSORS)} periods, start: {START}), spatial folds of 03; the same models found in: '
      + ', '.join(f'{label} ({what}) {n_compare[label]}' for label, (what, _) in COMPARE.items())
      + '; comparison: ' + ('; '.join(f'{label} R2 {v:.3f}' for label, v in REFS) or 'no runs of 05 or 07 found'))
if OOF is not None:
    print(f'oof_predictions.csv: {len(OOF)} test clusters, models: {[k for k in MODELS if k in OOF]}')
display(per_fold[FOLD_COLS].round(3))
results.round(3)
````
